```python
import geopandas as gpd
import matplotlib.pyplot as plt
import contextily as ctx
import pandas as pd
from shapely.geometry import MultiPolygon, Point
from matplotlib_scalebar.scalebar import ScaleBar
import matplotlib.gridspec as gridspec
import matplotlib.colors as colors
import numpy as np
import matplotlib as mpl
mpl.rcParams.update({'font.size': 8})

# Paths
shape_dir = "./map/1910map/shape"
gadm_path = "/Users/shwang/sandbox/data/borders/gadm36.shp"
depth_csv_path = "/Users/shwang/sandbox/shanghai/update2/fig124a3d.csv"
xyz1_path = "/Users/shwang/sandbox/shanghai/update2/fig3a.xyz"
xyz2_path = "/Users/shwang/sandbox/shanghai/update2/fig3b.xyz"
xyz3_path = "/Users/shwang/sandbox/shanghai/update2/fig3c.xyz"
rain_path = "./xujiahui_aug2031_daily_rain.csv"
river_mask_path = "/Users/shwang/sandbox/shanghai/river_mask.xyz"

# Load rainfall data
rain = pd.read_csv(rain_path, sep=",")
rain["date"] = pd.to_datetime(dict(year=rain.Year, month=rain.Mon, day=rain.Day))
rain["daily_mm"] = rain["daily_acc_10thmm"] / 10.0
rain["md"] = rain["date"].dt.strftime("%m-%d")

# Subset for Aug 20 – Sept 5
mask = (rain["date"].dt.month == 8) & (rain["date"].dt.day >= 20) | (rain["date"].dt.month == 9) & (rain["date"].dt.day <= 5)
rain_subset = rain.loc[mask].copy()
ordered_days = sorted(rain_subset["md"].unique(), key=lambda x: pd.to_datetime("1900-" + x))
rain_subset["md"] = pd.Categorical(rain_subset["md"], categories=ordered_days, ordered=True)

# Split 1905 from other years
rain_1905 = rain_subset[rain_subset["Year"] == 1905]
rain_clim = rain_subset[rain_subset["Year"] != 1905]

# Load historical map layers
圩塘 = gpd.read_file(f"{shape_dir}/圩塘.shp")
海岸线 = gpd.read_file(f"{shape_dir}/海岸线.shp")
钦公塘 = gpd.read_file(f"{shape_dir}/钦公塘.shp")
雍正石塘 = gpd.read_file(f"{shape_dir}/雍正石塘.shp")

# Load GADM and extract Shanghai
gadm = gpd.read_file(gadm_path)
sh_parts = gadm[(gadm["NAME_0"] == "China") & (gadm["NAME_2"] == "Shanghai")]
sh_dissolved = sh_parts.dissolve()
geom = sh_dissolved.geometry.iloc[0]
if isinstance(geom, MultiPolygon):
    main_poly = max(geom.geoms, key=lambda p: p.area)
    shanghai_main = gpd.GeoDataFrame(geometry=[main_poly], crs=sh_dissolved.crs)
else:
    shanghai_main = sh_dissolved

# Inland border
other_regions = gadm[(gadm["NAME_0"] == "China") & (gadm["NAME_2"] != "Shanghai")]
china_other = other_regions.dissolve()
shanghai_boundary = gpd.GeoDataFrame(geometry=shanghai_main.boundary, crs=shanghai_main.crs)
china_other_boundary = gpd.GeoDataFrame(geometry=china_other.boundary, crs=china_other.crs)
inland_border = gpd.overlay(shanghai_boundary, china_other_boundary, how="intersection")

# Load point data

river_mask_df = pd.read_csv(river_mask_path, sep='\s+', header=None, names=["lon", "lat", "depth"])
river_mask_gdf = gpd.GeoDataFrame(river_mask_df, geometry=gpd.points_from_xy(river_mask_df.lon, river_mask_df.lat), crs="EPSG:4326")
river_coords = set(zip(river_mask_df["lon"], river_mask_df["lat"]))

depth_df = pd.read_csv(depth_csv_path, header=None, names=["lon", "lat", "depth"])
depth_df = depth_df[~depth_df[["lon", "lat"]].apply(tuple, axis=1).isin(river_coords)]
depth_gdf = gpd.GeoDataFrame(depth_df, geometry=gpd.points_from_xy(depth_df.lon, depth_df.lat), crs="EPSG:4326")

xyz1_df = pd.read_csv(xyz1_path, sep='\s+', header=None, names=["lon", "lat", "depth"])
xyz1_df = xyz1_df[~xyz1_df[["lon", "lat"]].apply(tuple, axis=1).isin(river_coords)]
xyz1_gdf = gpd.GeoDataFrame(xyz1_df, geometry=gpd.points_from_xy(xyz1_df.lon, xyz1_df.lat), crs="EPSG:4326")

xyz2_df = pd.read_csv(xyz2_path, sep='\s+', header=None, names=["lon", "lat", "depth"])
xyz2_df = xyz2_df[~xyz2_df[["lon", "lat"]].apply(tuple, axis=1).isin(river_coords)]
xyz2_gdf = gpd.GeoDataFrame(xyz2_df, geometry=gpd.points_from_xy(xyz2_df.lon, xyz2_df.lat), crs="EPSG:4326")

xyz3_df = pd.read_csv(xyz3_path, sep='\s+', header=None, names=["lon", "lat", "depth"])
xyz3_df = xyz3_df[~xyz3_df[["lon", "lat"]].apply(tuple, axis=1).isin(river_coords)]
xyz3_gdf = gpd.GeoDataFrame(xyz3_df, geometry=gpd.points_from_xy(xyz3_df.lon, xyz3_df.lat), crs="EPSG:4326")

# Reproject to EPSG:3857
for layer in [圩塘, 海岸线, 钦公塘, 雍正石塘, shanghai_main, inland_border, depth_gdf, xyz1_gdf, xyz2_gdf,xyz3_gdf]:
    layer.to_crs(epsg=3857, inplace=True)

# Create figure and gridspec
fig = plt.figure(figsize=(7, 9))
gs = gridspec.GridSpec(nrows=4, ncols=2, height_ratios=[.85, 0.15, 1, 1], width_ratios=[1, 1])
ax1 = fig.add_subplot(gs[0, :])  # Top row
ax2 = fig.add_subplot(gs[2, 0])
ax3 = fig.add_subplot(gs[2, 1])
ax4 = fig.add_subplot(gs[3, 0])
ax5 = fig.add_subplot(gs[3, 1])
gs.update(wspace=0.05, hspace=0.05)

## ax1: Bar plot of 1905 rainfall with non-parametric return level shading

# Load and preprocess rainfall data
df = pd.read_csv("./xujiahui_aug2031_daily_rain.csv")

# Step 1: Annual maximum daily rainfall (for non-parametric estimation)
df["date"] = pd.to_datetime(dict(year=df.Year, month=df.Mon, day=df.Day))
df["daily_mm"] = df["daily_acc_10thmm"] / 10.0
df["year"] = df["date"].dt.year
annual_max = df.groupby("year")["daily_mm"].max().dropna().sort_values(ascending=False)
n_years = len(annual_max)
ranks = np.arange(1, n_years + 1)
empirical_return_periods = (n_years + 1) / ranks

# Step 2: Interpolate non-parametric return levels
target_T = [1, 2, 5]
return_levels = np.interp(target_T, np.sort(empirical_return_periods), np.sort(annual_max.values))
r1, r2, r5 = return_levels

# Step 3: Plot 1905 rainfall bars + shaded return level bands

# Prepare 1905 rainfall data
rain_1905_sorted = rain_1905.sort_values("md")
rain_1905_series = rain_1905_sorted.set_index("md").loc[ordered_days]["daily_mm"].values
x = pd.to_datetime(["1905-" + d for d in ordered_days])

# Colors
color_bar = "#1f77b4"
color_0_1 = 'none'
color_1_2 = "#c6dbef"     # Light blue
color_2_5 = "#6baed6"    # Medium blue
color_5plus = "#2171b5"  # Dark blue

# Horizontal lines for return levels
ax1.axhline(r2, color="#6baed6", linestyle="--", linewidth=2)
ax1.axhline(r5, color="#2171b5", linestyle="--", linewidth=2)
# Annotate return levels at Aug 19
ax1.text(pd.to_datetime("1905-08-18"), r2, "2-yr rainfall\nreturn period", color="#6baed6", fontsize=8, ha='center',va='center',linespacing=1.6)
ax1.text(pd.to_datetime("1905-08-18"), r5, "5-yr rainfall\nreturn period", color="#2171b5", fontsize=8, ha='center',va='center',linespacing=1.6)

# Bar plot
ax1.bar(x, rain_1905_series, color=color_bar, label="Rainfall")

# Formatting
ax1.set_ylim([0, 100])
ax1.tick_params(axis='y', colors=color_bar)
ax1.set_ylabel("Daily Rainfall (mm)", color=color_bar)
ax1.set_xlim(pd.to_datetime("1905-08-16"), pd.to_datetime("1905-09-03"))
# Customize x-ticks: show labels only from 08-20 to 09-03
tick_dates = pd.date_range("1905-08-20", "1905-09-03")
ax1.set_xticks(tick_dates)
ax1.set_xticklabels([d.strftime("%m-%d") for d in tick_dates], rotation=45)

# Load tide data
# tide_df = pd.read_csv("./tide_wusongkou.csv")  # Update with full path if needed
tide_df = pd.read_csv("./wusongkou.csv")  # Update with full path if needed
tide_df.columns = tide_df.columns.str.strip()  # Remove any whitespace in headers
# First, fix the year manually to make it four-digit starting with 1900
def fix_year(d):
    parts = d.split()
    date_parts = parts[0].split('/')
    if len(date_parts[2]) == 2:
        date_parts[2] = '19' + date_parts[2]
    return '/'.join(date_parts) + ' ' + parts[1]

# Apply to the date column
tide_df["fixed_date"] = tide_df["date and time"].apply(fix_year)

# Now safely parse as datetime
tide_df["datetime"] = pd.to_datetime(tide_df["fixed_date"], format="%m/%d/%Y %H:%M")
tide_df["depth_m"] = tide_df["water depth (points) (m)"]
tide_mask = (tide_df["datetime"] >= "1905-08-20") & (tide_df["datetime"] <= "1905-09-05 23:59")
tide_plot = tide_df.loc[tide_mask]

# Plot tide on right y-axis
ax1_tide = ax1.twinx()
ax1_tide.plot(tide_plot["datetime"], tide_plot["depth_m"], color="#d62728", linewidth=1.5, label="Water Level")

ax1_tide.axhline(5.4, color="#ff7f0e", linestyle="--", linewidth=2)
ax1_tide.text(pd.to_datetime("1905-08-18"), 5.4, "10-yr water level\nreturn period", color="#ff7f0e", fontsize=8, ha='center',va='center',linespacing=1.6)

ax1_tide.tick_params(axis='y', colors="#d62728")
ax1_tide.set_ylabel("Water Level (m)", color="#d62728")

ax1_tide.set_xticks(tide_plot["datetime"])
ax1_tide.set_xticklabels(tide_plot["datetime"], rotation=45)
import matplotlib.dates as mdates
ax1_tide.xaxis.set_major_locator(mdates.DayLocator())
ax1_tide.xaxis.set_major_formatter(mdates.DateFormatter('%m-%d'))
ax1_tide.set_ylim([0,7])
# plt.setp(ax1.get_xticklabels(), rotation=45)
ax1_tide.set_xlim(pd.to_datetime("1905-08-16"), pd.to_datetime("1905-09-03"))
# Customize x-ticks: show labels only from 08-20 to 09-03
tick_dates = pd.date_range("1905-08-20", "1905-09-03")
ax1_tide.set_xticks(tick_dates)
ax1_tide.set_xticklabels([d.strftime("%m-%d") for d in tick_dates], rotation=45)

# Combine legends from both axes
lines_1, labels_1 = ax1.get_legend_handles_labels()
lines_2, labels_2 = ax1_tide.get_legend_handles_labels()
from matplotlib.lines import Line2D
from matplotlib.patches import Patch

custom_handles = [
    Patch(facecolor=color_bar, edgecolor='none', label='Rainfall'),
    Line2D([0], [0], color="#d62728", lw=1.5, label='Water/Level')
]

ax1.legend(custom_handles, [h.get_label() for h in custom_handles], loc="lower left", handlelength=1.5, handleheight=1,frameon=False)

# Add annotation for Predecessor Rain #1
ax1.annotate("Predecessor\nRain #1",
             xy=(pd.to_datetime("1905-08-21"), 83.6), xycoords='data',
             xytext=(0, 5), textcoords='offset points',
             ha='center', va='bottom',
             arrowprops=dict(arrowstyle='-|>', facecolor='black', lw=0.8),
             fontsize=8, color='black')

# Add annotation for Predecessor Rain #2
ax1.annotate("Predecessor\nRain #2",
             xy=(pd.to_datetime("1905-08-28"), 83.6), xycoords='data',
             xytext=(0, 5), textcoords='offset points',
             ha='center', va='bottom',
             arrowprops=dict(arrowstyle='-|>', facecolor='black', lw=0.8),
             fontsize=8, color='black')

# Add annotation for Typhoon Impact (spanning 09-01 to 09-02)
ax1.annotate("Typhoon\nImpact",
             xy=(pd.to_datetime("1905-09-01 12:00"), 83.6), xycoords='data',
             xytext=(0, 5), textcoords='offset points',
             ha='center', va='bottom',
             arrowprops=dict(arrowstyle='-|>', facecolor='black', lw=0.8),
             fontsize=8, color='black')


# ax2-5

# Shared colormap
cmap = plt.cm.Blues
norm = colors.Normalize(vmin=0, vmax=1.2)

for ax_tmp,data_tmp in [[ax2,xyz1_gdf],[ax3,xyz2_gdf],[ax4,xyz3_gdf],[ax5,depth_gdf]]:
    圩塘.plot(ax=ax_tmp, color='darkgoldenrod', edgecolor='none')
    钦公塘.plot(ax=ax_tmp, color='darkgoldenrod', edgecolor='none', lw=2)
    雍正石塘.plot(ax=ax_tmp, edgecolor='slategray', linewidth=3)
    海岸线.plot(ax=ax_tmp, edgecolor='black', linewidth=2, linestyle='dashed')
    inland_border.plot(ax=ax_tmp, edgecolor='black', linewidth=0.8)
    ctx.add_basemap(ax_tmp, source=ctx.providers.CartoDB.PositronNoLabels,attribution=False)

    shanghai_main_tmp = shanghai_main.to_crs(epsg=3857)

    # Filter depth points inside Shanghai
    depth_within = gpd.sjoin(data_tmp, shanghai_main_tmp, how='inner', predicate='within')    
    hb = ax_tmp.hexbin(depth_within.geometry.x, depth_within.geometry.y,
                   C=depth_within["depth"], reduce_C_function=np.mean,
                   gridsize=70, cmap=cmap,norm=norm, edgecolors='none')
    hb.set_antialiased(False)
    
# Create a shared colorbar from ax3 to ax5
cbar_ax = fig.add_axes([0.92, ax5.get_position().y0, 0.015, ax3.get_position().y1 - ax5.get_position().y0])
sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
sm.set_array([])
cbar = fig.colorbar(sm, cax=cbar_ax)
cbar.set_label("Water Depth (m)")


from pyproj import Transformer

# Transformer from web mercator to lat/lon
transformer = Transformer.from_crs("EPSG:3857", "EPSG:4326", always_xy=True)

# Step 1: Get Shanghai mainland bounds in lat/lon (EPSG:4326)
bounds4326 = shanghai_main.to_crs(epsg=4326).total_bounds  # [minx, miny, maxx, maxy]

# Step 2: Transform bounds to EPSG:3857
transformer_to_3857 = Transformer.from_crs("EPSG:4326", "EPSG:3857", always_xy=True)
minx_3857, miny_3857 = transformer_to_3857.transform(bounds4326[0]-.05, bounds4326[1]-.05)
maxx_3857, maxy_3857 = transformer_to_3857.transform(bounds4326[2]+.05, bounds4326[3]+.05)

# Custom axes formatting for map subplots
axes_grid = [ax2, ax3, ax4, ax5]
show_x = [False, False, True, True]  # Only bottom row (ax4, ax5)
show_y = [True, False, True, False]  # Only left column (ax2, ax4)

for i, ax in enumerate(axes_grid):

    # Generate 0.2° interval ticks in lat/lon
    x_deg_ticks = np.arange(np.floor(bounds4326[0]*5)/5, np.ceil(bounds4326[2]*5)/5 + 0.01, 0.4)
    y_deg_ticks = np.arange(np.floor(bounds4326[1]*5)/5, np.ceil(bounds4326[3]*5)/5 + 0.01, 0.4)

    # Convert lat/lon ticks to EPSG:3857
    x_ticks_merc = [transformer_to_3857.transform(x, bounds4326[1])[0] for x in x_deg_ticks]
    y_ticks_merc = [transformer_to_3857.transform(bounds4326[0], y)[1] for y in y_deg_ticks]

    # Format tick labels
    x_labels = [f"{abs(round(x,2))}°{'E' if x >= 0 else 'W'}" for x in x_deg_ticks]
    y_labels = [f"{abs(round(y,2))}°{'N' if y >= 0 else 'S'}" for y in y_deg_ticks]

    ax.set_xticks(x_ticks_merc)
    ax.set_xticklabels([])
    ax.set_yticks(y_ticks_merc)
    ax.set_yticklabels([])
    
    # Apply tick positions and labels conditionally
    if show_x[i]:
        ax.set_xticklabels(x_labels)
    # else:
    #     ax.set_xticklabels([])

    if show_y[i]:
        ax.set_yticklabels(y_labels)
    # else:
    #     ax.set_yticklabels([])

    # Set consistent spatial extent
    ax.set_xlim(minx_3857, maxx_3857)
    ax.set_ylim(miny_3857, maxy_3857)

# Add bold subplot labels
labels = ['(a)', '(b)', '(c)', '(d)', '(e)']
axes_all = [ax1, ax2, ax3, ax4, ax5]

for ax, label in zip(axes_all, labels):
    ax.text(0.01, 0.98, label, transform=ax.transAxes,
            fontsize=10, fontweight='bold', va='top', ha='left')

# Add top-right corner titles for ax2–ax5
map_titles = ["Predecessor\nRain #1", "Predecessor\nRain #2", 
              "Before\nTyphoon Impact", "After\nTyphoon Impact"]
map_axes = [ax2, ax3, ax4, ax5]

for ax, title in zip(map_axes, map_titles):
    ax.text(0.99, 0.98, title, transform=ax.transAxes,
            fontsize=8, ha='right', va='top')

# Legend handles for ax5
from matplotlib.patches import Patch
from matplotlib.lines import Line2D

legend_elements = [
    Line2D([0], [0], color='darkgoldenrod', linewidth=2, label='Stone Embankment'),
    Line2D([0], [0], color='slategray', linewidth=3, label='Stone Embankment'),
    Line2D([0], [0], color='black', linestyle='dashed', linewidth=2, label='Coastline in 1950'),
    Line2D([0], [0], color='black', linewidth=0.8, label='Shanghai Border')
]

# Place the legend below ax4 and ax5, outside the figure area
fig.legend(handles=legend_elements, loc='lower center',
           bbox_to_anchor=(0.55, 0.05), ncol=4, fontsize=8, frameon=False)

plt.show()
```

```python
import geopandas as gpd
import matplotlib.pyplot as plt
import contextily as ctx
import pandas as pd
from shapely.geometry import MultiPolygon, Point
from matplotlib_scalebar.scalebar import ScaleBar
import matplotlib.gridspec as gridspec
import matplotlib.colors as colors
import numpy as np
import matplotlib as mpl
mpl.rcParams.update({'font.size': 8})

# Paths
shape_dir = "./map/1910map/shape"
gadm_path = "/Users/shwang/sandbox/data/borders/gadm36.shp"
depth_csv_path = "/Users/shwang/sandbox/shanghai/update2/fig124a3d.csv"
norain1_path = "/Users/shwang/sandbox/shanghai/update2/norain1/text/t358.xyz"
norain2_path = "/Users/shwang/sandbox/shanghai/update2/norain2/text/t358.xyz"
stone2earth_csv_path = "/Users/shwang/sandbox/shanghai/update2/fig4b.csv"
lowtide_csv_path = "/Users/shwang/sandbox/shanghai/update2/fig2b.csv"
river_mask_path = "/Users/shwang/sandbox/shanghai/river_mask.xyz"

# Load historical map layers
圩塘 = gpd.read_file(f"{shape_dir}/圩塘.shp")
海岸线 = gpd.read_file(f"{shape_dir}/海岸线.shp")
钦公塘 = gpd.read_file(f"{shape_dir}/钦公塘.shp")
雍正石塘 = gpd.read_file(f"{shape_dir}/雍正石塘.shp")

# Load GADM and extract Shanghai
gadm = gpd.read_file(gadm_path)
sh_parts = gadm[(gadm["NAME_0"] == "China") & (gadm["NAME_2"] == "Shanghai")]
sh_dissolved = sh_parts.dissolve()
geom = sh_dissolved.geometry.iloc[0]
if isinstance(geom, MultiPolygon):
    main_poly = max(geom.geoms, key=lambda p: p.area)
    shanghai_main = gpd.GeoDataFrame(geometry=[main_poly], crs=sh_dissolved.crs)
else:
    shanghai_main = sh_dissolved

# Inland border
other_regions = gadm[(gadm["NAME_0"] == "China") & (gadm["NAME_2"] != "Shanghai")]
china_other = other_regions.dissolve()
shanghai_boundary = gpd.GeoDataFrame(geometry=shanghai_main.boundary, crs=shanghai_main.crs)
china_other_boundary = gpd.GeoDataFrame(geometry=china_other.boundary, crs=china_other.crs)
inland_border = gpd.overlay(shanghai_boundary, china_other_boundary, how="intersection")

# Load point data

river_mask_df = pd.read_csv(river_mask_path, sep='\s+', header=None, names=["lon", "lat", "depth"])
river_mask_gdf = gpd.GeoDataFrame(river_mask_df, geometry=gpd.points_from_xy(river_mask_df.lon, river_mask_df.lat), crs="EPSG:4326")
river_coords = set(zip(river_mask_df["lon"], river_mask_df["lat"]))

depth_df = pd.read_csv(depth_csv_path, header=None, names=["lon", "lat", "depth"])
depth_df = depth_df[~depth_df[["lon", "lat"]].apply(tuple, axis=1).isin(river_coords)]
depth_gdf = gpd.GeoDataFrame(depth_df, geometry=gpd.points_from_xy(depth_df.lon, depth_df.lat), crs="EPSG:4326")

stone2earth_df = pd.read_csv(stone2earth_csv_path, header=None, names=["lon", "lat", "depth"])
stone2earth_df = stone2earth_df[~stone2earth_df[["lon", "lat"]].apply(tuple, axis=1).isin(river_coords)]
stone2earth_gdf = gpd.GeoDataFrame(stone2earth_df, geometry=gpd.points_from_xy(stone2earth_df.lon, stone2earth_df.lat), crs="EPSG:4326")

lowtide_df = pd.read_csv(lowtide_csv_path, header=None, names=["lon", "lat", "depth"])
lowtide_df = lowtide_df[~lowtide_df[["lon", "lat"]].apply(tuple, axis=1).isin(river_coords)]
lowtide_gdf = gpd.GeoDataFrame(lowtide_df, geometry=gpd.points_from_xy(lowtide_df.lon, lowtide_df.lat), crs="EPSG:4326")

norain1_df = pd.read_csv(norain1_path, sep='\s+', header=None, names=["lon", "lat", "depth"])
norain1_df = norain1_df[~norain1_df[["lon", "lat"]].apply(tuple, axis=1).isin(river_coords)]
norain1_gdf = gpd.GeoDataFrame(norain1_df, geometry=gpd.points_from_xy(norain1_df.lon, norain1_df.lat), crs="EPSG:4326")

norain2_df = pd.read_csv(norain2_path, sep='\s+', header=None, names=["lon", "lat", "depth"])
norain2_df = norain2_df[~norain2_df[["lon", "lat"]].apply(tuple, axis=1).isin(river_coords)]
norain2_gdf = gpd.GeoDataFrame(norain2_df, geometry=gpd.points_from_xy(norain2_df.lon, norain2_df.lat), crs="EPSG:4326")

# Reproject to EPSG:3857
for layer in [圩塘, 海岸线, 钦公塘, 雍正石塘, shanghai_main, inland_border, depth_gdf, stone2earth_gdf, lowtide_gdf,norain1_gdf,norain2_gdf]:
    layer.to_crs(epsg=3857, inplace=True)

# Create figure and gridspec
fig = plt.figure(figsize=(7, 6),dpi=100)
gs = gridspec.GridSpec(nrows=2, ncols=2, height_ratios=[1, 1], width_ratios=[1, 1])
ax1 = fig.add_subplot(gs[0, 0]) 
ax2 = fig.add_subplot(gs[0, 1])
ax3 = fig.add_subplot(gs[1, 0])
ax4 = fig.add_subplot(gs[1, 1])
gs.update(wspace=0.05, hspace=0.05)

# ax1-4
import matplotlib.colors as mcolors
def shuai_colormap2():
    C1 = np.asarray([170,219,30])/255 # green
    C2 = np.asarray([2,7,93])/255 # dark blue
    C3 = np.asarray([0,183,235])/255 # cyan
    C4 = np.asarray([179,178,218])/255 # light purple
    C5 = np.asarray([255,255,255])/255 # white
    C6 = np.asarray([255,0,0])/255 # red
    C7 = np.asarray([255,233,0])/255 # yellow
    C8 = np.asarray([255,240,240])/255 # mistyrose
    C9 = np.asarray([240, 255, 255])/255 # azure
    CUTS = 10
    tmp1 = np.asarray([np.linspace(C1[0],C2[0],CUTS),np.linspace(C1[1],C2[1],CUTS),np.linspace(C1[2],C2[2],CUTS)]).transpose()[:-1]
    CUTS = 8
    tmp2 = np.asarray([np.linspace(C2[0],C3[0],CUTS),np.linspace(C2[1],C3[1],CUTS),np.linspace(C2[2],C3[2],CUTS)]).transpose()[:-1]
    # CUTS = 3
    # tmp3 = np.asarray([np.linspace(C3[0],C4[0],CUTS),np.linspace(C3[1],C4[1],CUTS),np.linspace(C3[2],C4[2],CUTS)]).transpose()[:-1]
    CUTS = 5
    tmp4 = np.asarray([np.linspace(C3[0],C9[0],CUTS),np.linspace(C3[1],C9[1],CUTS),np.linspace(C3[2],C9[2],CUTS)]).transpose()
    CUTS = 10
    tmp5 = np.asarray([np.linspace(C8[0],C6[0],CUTS),np.linspace(C8[1],C6[1],CUTS),np.linspace(C8[2],C6[2],CUTS)]).transpose()[:-1]
    CUTS = 12
    tmp6 = np.asarray([np.linspace(C6[0],C7[0],CUTS),np.linspace(C6[1],C7[1],CUTS),np.linspace(C6[2],C7[2],CUTS)]).transpose()
    colors = np.concatenate([tmp1,tmp2,tmp4,tmp5,tmp6])
    cmap = mcolors.LinearSegmentedColormap.from_list('my_colormap', colors)
    cmaplist = [cmap(i) for i in range(cmap.N)]
    cmap = mcolors.LinearSegmentedColormap.from_list('Custom cmap', cmaplist, cmap.N)
    return cmap.reversed()

# Shared colormap
cmap = shuai_colormap2()
norm = colors.Normalize(vmin=-.5, vmax=.5)

for ax_tmp,data_tmp in [[ax1,norain1_gdf],[ax2,norain2_gdf],[ax3,lowtide_gdf],[ax4,stone2earth_gdf]]:
    圩塘.plot(ax=ax_tmp, color='darkgoldenrod', edgecolor='none')
    钦公塘.plot(ax=ax_tmp, color='darkgoldenrod', edgecolor='none', lw=2)
    if ax_tmp == ax4:
        雍正石塘.plot(ax=ax_tmp, edgecolor='darkgoldenrod', lw=2)
    else:
        雍正石塘.plot(ax=ax_tmp, edgecolor='slategray', lw=3)
    海岸线.plot(ax=ax_tmp, edgecolor='black', linewidth=2, linestyle='dashed')
    inland_border.plot(ax=ax_tmp, edgecolor='black', linewidth=0.8)
    ctx.add_basemap(ax_tmp, source=ctx.providers.CartoDB.PositronNoLabels,attribution=False)
    
    shanghai_main_tmp = shanghai_main.to_crs(epsg=3857)

    # Filter depth points inside Shanghai
    depth_within      = gpd.sjoin(data_tmp , shanghai_main_tmp, how='inner', predicate='within')    
    depth_within_ctrl = gpd.sjoin(depth_gdf, shanghai_main_tmp, how='inner', predicate='within')    
    
############
    from scipy.stats import binned_statistic_2d

    # Set consistent grid definition
    gridsize = 70
    
    # Combine all coordinates to define a common spatial range
    x_all = np.concatenate([depth_within.geometry.x.values, depth_within_ctrl.geometry.x.values])
    y_all = np.concatenate([depth_within.geometry.y.values, depth_within_ctrl.geometry.y.values])
    
    x_range = [x_all.min(), x_all.max()]
    y_range = [y_all.min(), y_all.max()]
    
    # Bin depth_within (e.g., post-typhoon)
    stat1, xedges, yedges, _ = binned_statistic_2d(
        depth_within.geometry.x, depth_within.geometry.y, depth_within["depth"],
        statistic='mean', bins=gridsize*1.5, range=[x_range, y_range]
    )
    
    # Bin depth_within_ctrl (e.g., pre-typhoon), using same edges
    stat2, _, _, _ = binned_statistic_2d(
        depth_within_ctrl.geometry.x, depth_within_ctrl.geometry.y, depth_within_ctrl["depth"],
        statistic='mean', bins=[xedges, yedges]
    )
    
    # Replace NaNs with 0
    stat1_filled = np.nan_to_num(stat1, nan=0)
    stat2_filled = np.nan_to_num(stat2, nan=0)
    
    # Compute difference
    diff = stat1_filled - stat2_filled
    
    # Prepare grid centers
    x_centers = (xedges[:-1] + xedges[1:]) / 2
    y_centers = (yedges[:-1] + yedges[1:]) / 2
    Xc, Yc = np.meshgrid(x_centers, y_centers)
    
    # Flatten
    x_flat = Xc.ravel()
    y_flat = Yc.ravel()
    z_flat = diff.T.ravel()

    # Plot hexbin difference
    hb = ax_tmp.hexbin(
        x_flat[abs(z_flat)>.01], y_flat[abs(z_flat)>.01], C=z_flat[abs(z_flat)>.01],
        gridsize=gridsize, cmap=cmap, norm=norm, reduce_C_function=np.mean,edgecolors='none'
    )

############

    
    hb.set_antialiased(False)
    
# Create a shared colorbar from ax3 to ax5
cbar_ax = fig.add_axes([0.92, ax4.get_position().y0, 0.015, ax2.get_position().y1 - ax4.get_position().y0])
sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
sm.set_array([])
cbar = fig.colorbar(sm, cax=cbar_ax)
cbar.set_label("Water Depth (m)")


from pyproj import Transformer

# Transformer from web mercator to lat/lon
transformer = Transformer.from_crs("EPSG:3857", "EPSG:4326", always_xy=True)

# Step 1: Get Shanghai mainland bounds in lat/lon (EPSG:4326)
bounds4326 = shanghai_main.to_crs(epsg=4326).total_bounds  # [minx, miny, maxx, maxy]

# Step 2: Transform bounds to EPSG:3857
transformer_to_3857 = Transformer.from_crs("EPSG:4326", "EPSG:3857", always_xy=True)
minx_3857, miny_3857 = transformer_to_3857.transform(bounds4326[0]-.05, bounds4326[1]-.05)
maxx_3857, maxy_3857 = transformer_to_3857.transform(bounds4326[2]+.05, bounds4326[3]+.05)

# Custom axes formatting for map subplots
axes_grid = [ax1, ax2, ax3, ax4]
show_x = [False, False, True, True]  # Only bottom row (ax4, ax5)
show_y = [True, False, True, False]  # Only left column (ax2, ax4)

for i, ax in enumerate(axes_grid):

    # Generate 0.2° interval ticks in lat/lon
    x_deg_ticks = np.arange(np.floor(bounds4326[0]*5)/5, np.ceil(bounds4326[2]*5)/5 + 0.01, 0.4)
    y_deg_ticks = np.arange(np.floor(bounds4326[1]*5)/5, np.ceil(bounds4326[3]*5)/5 + 0.01, 0.4)

    # Convert lat/lon ticks to EPSG:3857
    x_ticks_merc = [transformer_to_3857.transform(x, bounds4326[1])[0] for x in x_deg_ticks]
    y_ticks_merc = [transformer_to_3857.transform(bounds4326[0], y)[1] for y in y_deg_ticks]

    # Format tick labels
    x_labels = [f"{abs(round(x,2))}°{'E' if x >= 0 else 'W'}" for x in x_deg_ticks]
    y_labels = [f"{abs(round(y,2))}°{'N' if y >= 0 else 'S'}" for y in y_deg_ticks]

    ax.set_xticks(x_ticks_merc)
    ax.set_xticklabels([])
    ax.set_yticks(y_ticks_merc)
    ax.set_yticklabels([])
    
    # Apply tick positions and labels conditionally
    if show_x[i]:
        ax.set_xticklabels(x_labels)
    # else:
    #     ax.set_xticklabels([])

    if show_y[i]:
        ax.set_yticklabels(y_labels)
    # else:
    #     ax.set_yticklabels([])

    # Set consistent spatial extent
    ax.set_xlim(minx_3857, maxx_3857)
    ax.set_ylim(miny_3857, maxy_3857)

# Add bold subplot labels
labels = ['(a)', '(b)', '(c)', '(d)']
axes_all = [ax1, ax2, ax3, ax4]

for ax, label in zip(axes_all, labels):
    ax.text(0.01, 0.98, label, transform=ax.transAxes,
            fontsize=10, fontweight='bold', va='top', ha='left')

# Add top-right corner titles for ax2–ax5
map_titles = ["No Rain #1 - CTRL", "No Rain #2 - CTRL", 
              "Low Tide - CTRL", "No Stone Enbankment - CTRL"]
map_axes = [ax1, ax2, ax3, ax4]

for ax, title in zip(map_axes, map_titles):
    ax.text(0.99, 0.98, title, transform=ax.transAxes,
            fontsize=8, ha='right', va='top')

# Legend handles for ax5
from matplotlib.patches import Patch
from matplotlib.lines import Line2D

legend_elements = [
    Line2D([0], [0], color='darkgoldenrod', linewidth=2, label='Earthen Embankment'),
    Line2D([0], [0], color='slategray', linewidth=3, label='Stone Embankment'),
    Line2D([0], [0], color='black', linestyle='dashed', linewidth=2, label='Coastline in 1950'),
    Line2D([0], [0], color='black', linewidth=0.8, label='Shanghai Border')
]

# Place the legend below ax4 and ax5, outside the figure area
fig.legend(handles=legend_elements, loc='lower center',
           bbox_to_anchor=(0.55, 0.02), ncol=4, fontsize=8, frameon=False)

plt.show()
```
