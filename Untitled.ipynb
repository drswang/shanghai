```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load and preprocess rainfall data
df = pd.read_csv("/Users/shwang/sandbox/shanghai/xujiahui_aug2031_daily_rain.csv")
df["date"] = pd.to_datetime(dict(year=df.Year, month=df.Mon, day=df.Day))
df["daily_mm"] = df["daily_acc_10thmm"] / 10.0
df["year"] = df["date"].dt.year

# Step 1: Get annual maximum daily rainfall
annual_max = df.groupby("year")["daily_mm"].max().dropna()
annual_max_sorted = np.sort(annual_max)[::-1]  # descending order

# Step 2: Assign empirical return periods
n = len(annual_max_sorted)
ranks = np.arange(1, n + 1)
empirical_return_periods = (n + 1) / ranks  # Weibull formula

# Step 3: Interpolate for desired return periods
desired_T = np.array([2,3,4,5,6,7,8,9,10])
rainfall_levels = np.interp(desired_T, np.sort(empirical_return_periods), np.sort(annual_max_sorted))

# Print results
for T, R in zip(desired_T, rainfall_levels):
    print(f"{T}-year event (non-parametric estimate): {R:.2f} mm")

# Optional: plot the return level curve
plt.figure(figsize=(6, 4))
plt.plot(empirical_return_periods, annual_max_sorted, marker='o', linestyle='-', label='Empirical')
plt.scatter(desired_T, rainfall_levels, color='red', label='Estimated Return Levels')
plt.xscale("log")
plt.xlabel("Return Period (years)")
plt.ylabel("Annual Max Daily Rainfall (mm)")
plt.title("Non-parametric Return Level Plot")
plt.grid(True, which="both", ls="--", alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import lognorm

# X range (positive values for lognormal)
x = np.linspace(0.2, 4, 1000)

# Distribution 1: Lognormal
sigma1, mu1 = 0.4, 0.0   # shape and location parameters
y1 = lognorm.pdf(x, sigma1, scale=np.exp(mu1))

# Distribution 2: Slightly higher mean, larger sigma, 20% lower frequency
sigma2, mu2 = 0.45, 0.1
y2 = 0.99 * lognorm.pdf(x, sigma2, scale=np.exp(mu2))

# Plot
plt.figure(figsize=(6, 4),dpi=600)
plt.plot(x, y1, label='Historical',lw=3)
plt.plot(x, y2, label='Future',lw=3)

# Add vertical line at x = 2
plt.axvline(x=2, color='gray', linestyle='--', linewidth=1)

# Add text near the vertical line
plt.text(2.05, max(y1.max(), y2.max()) * 0.8, 'Tipping point', rotation=90,
         verticalalignment='center', color='gray', fontsize=16)

# Shade the area between y1 and y2 to the right of x = 2
mask = x >= 2
plt.fill_between(x[mask], y1[mask], y2[mask], color='red', alpha=0.8,label="Risk increase")

# Remove axis ticks and labels
plt.xticks([])
plt.yticks([])
plt.tick_params(left=False, bottom=False)

# Axis titles
plt.xlabel('Hazard Severity')
plt.ylabel('Hazard frequency')
# Remove legend and top/right frame lines
plt.legend()
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.ylim(0, max(y1.max(), y2.max())*1.05)
plt.xlim([.3,3.6])

plt.tight_layout()
plt.show()
```
