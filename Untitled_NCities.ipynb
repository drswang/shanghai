```python
"""
Data Processing Cell - Load all shared data for Figures 1-6
============================================================
This cell loads:
- Geographic data (GADM boundaries, historical shapefiles)
- Rainfall data
- Tide data (observed and astronomical)
- River masks
- Flood depth data from all scenarios (ctrl, lowtide, norain1, norain2, nostone)
- Interpolates data onto hex grid for consistent area/volume calculations
- Computes max depth, duration, and area time series for all scenarios
"""

import numpy as np
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import matplotlib.colors as mcolors
import matplotlib.gridspec as gridspec
from matplotlib.patches import Patch
from matplotlib.lines import Line2D
from matplotlib_scalebar.scalebar import ScaleBar
from shapely.geometry import MultiPolygon, LineString, Point, Polygon
import contextily as ctx
from pyproj import Transformer
import os
import matplotlib as mpl
from scipy.interpolate import griddata
mpl.rcParams.update({'font.size': 8})

# ============================================================================
# PATHS
# ============================================================================
base_dir = "/Users/shwang/sandbox/shanghai"
rain_path = f"{base_dir}/xujiahui_aug2031_daily_rain.csv"
# Note: Tide data paths are now handled in the NetCDF extraction cell
# wusong_obs_path and wusong_tide_path are no longer needed
river_dir = f"{base_dir}/市管河道"
shape_dir = f"{base_dir}/map/1910map/shape"
gadm_path = os.path.expanduser("~/sandbox/data/borders/gadm36.shp")

# Scenario directories
# ctrl: Control scenario (baseline with all embankments)
# lowtide: Neap tide scenario
# norain1, norain2: No rainfall scenarios
# nostone: No stone embankments (earthen embankments only)
scenario_dirs = {
    "ctrl": f"{base_dir}/update2/ctrl/text",
    "lowtide": f"{base_dir}/update2/lowtide/text",
    "norain1": f"{base_dir}/update2/norain1/text",
    "norain2": f"{base_dir}/update2/norain2/text",
    "nostone": f"{base_dir}/update2/nostone/text"
}

# Simulation parameters
SIM_START = pd.Timestamp("1905-08-19 01:00:00")
N_TIMESTEPS = 358
DEPTH_THRESHOLD = 0.3  # meters for inundation
HEX_GRIDSIZE = 100  # Same as visualization gridsize

# Pre-surge timestamp: 1905-08-31 18:00
PRE_SURGE_TIME = pd.Timestamp("1905-08-19 02:00:00")#pd.Timestamp("1905-08-31 18:00:00")
PRE_SURGE_TIMESTEP = int((PRE_SURGE_TIME - SIM_START).total_seconds() / 3600) + 1
print(f"Pre-surge timestep: {PRE_SURGE_TIMESTEP} (corresponds to {PRE_SURGE_TIME})")

# ============================================================================
# LOAD RAINFALL DATA
# ============================================================================
print("\nLoading rainfall data...")
rain = pd.read_csv(rain_path, sep=",")
rain["date"] = pd.to_datetime(dict(year=rain.Year, month=rain.Mon, day=rain.Day))
rain["daily_mm"] = rain["daily_acc_10thmm"] / 10.0
rain["md"] = rain["date"].dt.strftime("%m-%d")

# Subset for Aug 20 – Sept 5
mask = (rain["date"].dt.month == 8) & (rain["date"].dt.day >= 20) | (rain["date"].dt.month == 9) & (rain["date"].dt.day <= 5)
rain_subset = rain.loc[mask].copy()
ordered_days = sorted(rain_subset["md"].unique(), key=lambda x: pd.to_datetime("1900-" + x))
rain_subset["md"] = pd.Categorical(rain_subset["md"], categories=ordered_days, ordered=True)
rain_subset = rain_subset[rain_subset["md"].between("08-20", "09-02")]
ordered_days = sorted(rain_subset["md"].unique(), key=lambda x: pd.to_datetime("1900-" + x))

# Split 1905 from other years
rain_1905 = rain_subset[rain_subset["Year"] == 1905]

# Annual max for return period estimation
df_rain = pd.read_csv(rain_path)
df_rain["date"] = pd.to_datetime(dict(year=df_rain.Year, month=df_rain.Mon, day=df_rain.Day))
df_rain["daily_mm"] = df_rain["daily_acc_10thmm"] / 10.0
df_rain["year"] = df_rain["date"].dt.year
annual_max = df_rain.groupby("year")["daily_mm"].max().dropna().sort_values(ascending=False)
n_years = len(annual_max)
ranks = np.arange(1, n_years + 1)
empirical_return_periods = (n_years + 1) / ranks

# Interpolate non-parametric return levels
target_T = [1, 2, 5]
return_levels = np.interp(target_T, np.sort(empirical_return_periods), np.sort(annual_max.values))
R1_LEVEL, R2_LEVEL, R5_LEVEL = return_levels
print(f"  Return levels: 1-yr={R1_LEVEL:.1f}mm, 2-yr={R2_LEVEL:.1f}mm, 5-yr={R5_LEVEL:.1f}mm")

# ============================================================================
# TIDE / WATER LEVEL DATA 
# ============================================================================
# NOTE: Tide data is now loaded from NetCDF files in the "Wusongkou Station Water Level Extraction" cell.
# This includes:
#   - tide_astro: Astronomical tide (no surge) from FlowFM_0000_his.nc
#   - tide_spring: Spring tide + surge from FlowFM_0000_his_890.nc
#   - tide_neap: Neap tide + surge from FlowFM_0000_his_890_7.nc
#   - wusong_simulated: Dict mapping scenarios to their water level DataFrames
print("\nTide data will be loaded from NetCDF files in the extraction cell...")

# ============================================================================
# LOAD GEOGRAPHIC DATA
# ============================================================================
print("\nLoading geographic data...")

gadm = gpd.read_file(gadm_path)
sh_parts = gadm[(gadm["NAME_0"] == "China") & (gadm["NAME_2"] == "Shanghai")]
sh_dissolved = sh_parts.dissolve()
geom = sh_dissolved.geometry.iloc[0]
if isinstance(geom, MultiPolygon):
    main_poly = max(geom.geoms, key=lambda p: p.area)
    shanghai_main = gpd.GeoDataFrame(geometry=[main_poly], crs=sh_dissolved.crs)
else:
    shanghai_main = sh_dissolved

other_regions = gadm[(gadm["NAME_0"] == "China") & (gadm["NAME_2"] != "Shanghai")]
china_other = other_regions.dissolve()
shanghai_boundary = gpd.GeoDataFrame(geometry=shanghai_main.boundary, crs=shanghai_main.crs)
china_other_boundary = gpd.GeoDataFrame(geometry=china_other.boundary, crs=china_other.crs)
inland_border = gpd.overlay(shanghai_boundary, china_other_boundary, how="intersection")

# Historical shapefiles
海岸线 = gpd.read_file(f"{shape_dir}/海岸线.shp")
圩塘 = gpd.read_file(f"{shape_dir}/圩塘.shp")
钦公塘 = gpd.read_file(f"{shape_dir}/钦公塘.shp")
雍正石塘 = gpd.read_file(f"{shape_dir}/雍正石塘.shp")

# Load 市管河道 shapefile (for visualization only)
市管河道 = gpd.read_file(f"{river_dir}/市管河道.shp")
if 市管河道.crs is None:
    市管河道 = 市管河道.set_crs("EPSG:4326")

print(f"  Shanghai boundary loaded")
print(f"  Historical shapefiles loaded (海岸线, 圩塘, 钦公塘, 雍正石塘)")

# ============================================================================
# CREATE RIVER/OCEAN MASK FROM INITIAL WATER DEPTH
# ============================================================================
# Use t1.xyz from ctrl scenario to identify river/ocean grid points
# Grid points with water depth > 3m at t=1 are considered river/ocean
print("\nCreating river/ocean mask from initial water depth...")
RIVER_OCEAN_DEPTH_THRESHOLD = .0  # meters

t1_path = f"{base_dir}/update2/ctrl/text/t1.xyz"
t1_df = pd.read_csv(t1_path, sep=r"\s+", header=None, names=["lon", "lat", "depth"])

# Find grid points with depth > 3m (river/ocean)
river_ocean_mask = t1_df[t1_df["depth"] > RIVER_OCEAN_DEPTH_THRESHOLD]
river_ocean_coords = set(zip(river_ocean_mask["lon"], river_ocean_mask["lat"]))

print(f"  Reference file: {t1_path}")
print(f"  Depth threshold: > {RIVER_OCEAN_DEPTH_THRESHOLD} m")
print(f"  River/ocean grid points identified: {len(river_ocean_coords)}")
print(f"  Total grid points in t1.xyz: {len(t1_df)}")
print(f"  Land grid points remaining: {len(t1_df) - len(river_ocean_coords)}")

# ============================================================================
# CREATE HEX GRID FOR INTERPOLATION
# ============================================================================
print("\nCreating hex grid for interpolation...")

# Transform Shanghai boundary to Web Mercator for hex grid creation
shanghai_main_4326 = shanghai_main.to_crs(epsg=4326)
bounds = shanghai_main_4326.total_bounds  # [minx, miny, maxx, maxy]

# Add buffer to bounds
buffer = 0.05
minx, miny, maxx, maxy = bounds[0] - buffer, bounds[1] - buffer, bounds[2] + buffer, bounds[3] + buffer

# Create hex grid in Web Mercator (EPSG:3857)
transformer_to_3857 = Transformer.from_crs("EPSG:4326", "EPSG:3857", always_xy=True)
transformer_to_4326 = Transformer.from_crs("EPSG:3857", "EPSG:4326", always_xy=True)

minx_3857, miny_3857 = transformer_to_3857.transform(minx, miny)
maxx_3857, maxy_3857 = transformer_to_3857.transform(maxx, maxy)

# ============================================================================
# FLAT-TOP HEXAGON TESSELLATION (shared vertices and sides)
# ============================================================================
# For flat-top hexagons with circumradius R (center → vertex):
# - Width (vertex → vertex): 2R
# - Height (flat → flat): sqrt(3) * R
# - Center spacing (x): 1.5R
# - Center spacing (y): sqrt(3) * R
# - Odd columns shift vertically by half a hex height to share edges.

hex_size = (maxx_3857 - minx_3857) / (HEX_GRIDSIZE * 1.5)  # Circumradius R
hex_width = 2 * hex_size
hex_height = np.sqrt(3) * hex_size
hex_horiz_spacing = 1.5 * hex_size
hex_vert_spacing = hex_height

# Create hex centers with exact shared edges (flat-top axial layout)
hex_centers_x = []
hex_centers_y = []
col = 0
while True:
    x = minx_3857 + hex_size + col * hex_horiz_spacing
    if x > maxx_3857 + hex_size:
        break
    y_offset = (hex_height / 2) if (col % 2 == 1) else 0
    row = 0
    while True:
        y = miny_3857 + hex_height / 2 + y_offset + row * hex_vert_spacing
        if y > maxy_3857 + hex_height:
            break
        hex_centers_x.append(x)
        hex_centers_y.append(y)
        row += 1
    col += 1

hex_centers_3857 = np.array(list(zip(hex_centers_x, hex_centers_y)))

# Convert hex centers to lat/lon for interpolation
hex_centers_4326 = np.array([transformer_to_4326.transform(x, y) for x, y in hex_centers_3857])
hex_lons = hex_centers_4326[:, 0]
hex_lats = hex_centers_4326[:, 1]

# Create hex polygons for area calculation - FLAT-TOP orientation
def create_hex_polygon(cx, cy, size):
    """Create a regular flat-top hexagon polygon centered at (cx, cy).
    Flat-top means horizontal top/bottom edges (angles 0, 60, 120, 180, 240, 300).
    """
    angles = np.array([0, 60, 120, 180, 240, 300]) * np.pi / 180
    points = [(cx + size * np.cos(a), cy + size * np.sin(a)) for a in angles]
    return Polygon(points)

hex_polygons = [create_hex_polygon(cx, cy, hex_size) for cx, cy in hex_centers_3857]
hex_gdf = gpd.GeoDataFrame(geometry=hex_polygons, crs="EPSG:3857")

# Filter hex cells within Shanghai boundary
shanghai_3857 = shanghai_main.to_crs(epsg=3857)
hex_gdf["hex_id"] = range(len(hex_gdf))
hex_within = gpd.sjoin(hex_gdf, shanghai_3857, how="inner", predicate="intersects")
valid_hex_ids = set(hex_within["hex_id"].values)

# Calculate area of each hex cell (in km²)
HEX_AREA_M2 = hex_gdf.iloc[0].geometry.area  # All hexes same size
HEX_AREA_KM2 = HEX_AREA_M2 / 1e6

print(f"  Hex grid size: {HEX_GRIDSIZE}")
print(f"  Hex cell size (radius): {hex_size:.1f} m")
print(f"  Hex cell area: {HEX_AREA_KM2:.4f} km² ({HEX_AREA_M2:.1f} m²)")
print(f"  Total hex cells: {len(hex_centers_3857)}")
print(f"  Hex cells within Shanghai: {len(valid_hex_ids)}")

# Store hex grid info for later use
hex_grid_info = {
    "centers_3857": hex_centers_3857,
    "centers_4326": hex_centers_4326,
    "lons": hex_lons,
    "lats": hex_lats,
    "polygons": hex_polygons,
    "gdf": hex_gdf,
    "valid_ids": valid_hex_ids,
    "area_km2": HEX_AREA_KM2,
    "area_m2": HEX_AREA_M2
}

# ============================================================================
# HELPER FUNCTIONS FOR DATA LOADING AND INTERPOLATION
# ============================================================================
def load_xyz_data(fpath):
    """Load XYZ file and remove river/ocean grid cells.
    
    River/ocean cells are identified as grid points with depth > 3m 
    in the initial timestep (t1.xyz from ctrl scenario).
    """
    df = pd.read_csv(fpath, sep=r"\s+", header=None, names=["lon", "lat", "depth"])
    df = df[~df[["lon", "lat"]].apply(tuple, axis=1).isin(river_ocean_coords)]
    return df

def interpolate_to_hex(df, value_col="depth"):
    """Assign mean value per hex by averaging all points within each hex."""
    pts = gpd.GeoDataFrame(
        df[[value_col]].copy(),
        geometry=gpd.points_from_xy(df.lon, df.lat),
        crs="EPSG:4326",
    ).to_crs(epsg=3857)

    joined = gpd.sjoin(
        pts,
        hex_gdf[["hex_id", "geometry"]],
        how="inner",
        predicate="within",
    )

    agg = joined.groupby("hex_id")[value_col].mean()
    interpolated = np.full(len(hex_gdf), np.nan)
    interpolated[agg.index] = agg.values
    return interpolated

def create_hex_gdf(values, value_col="depth"):
    """Create GeoDataFrame with hex polygons and interpolated values."""
    gdf = hex_gdf.copy()
    gdf[value_col] = values
    # Filter to valid hex cells within Shanghai
    gdf = gdf[gdf["hex_id"].isin(valid_hex_ids)].copy()
    return gdf

# ============================================================================
# PROCESS ALL SCENARIOS - COMPUTE MAX DEPTH, DURATION, AREA TIME SERIES
# ============================================================================
print("\n" + "="*60)
print("PROCESSING FLOOD DATA FOR ALL SCENARIOS (HEX GRID INTERPOLATION)")
print("="*60)

# Storage for all scenario data
scenario_data = {}

for scenario_name, xyz_dir in scenario_dirs.items():
    print(f"\n--- Processing {scenario_name} ---")

    # Initialize storage for hex-based calculations
    n_hex = len(hex_lons)
    max_depth_hex = np.full(n_hex, np.nan)
    hours_above_hex = np.zeros(n_hex)
    hours_above_presurge_hex = np.zeros(n_hex)
    area_timeseries = []
    stored_water_timeseries = []

    for t in range(1, N_TIMESTEPS + 1):
        if t % 50 == 0:
            print(f"  Timestep {t}/{N_TIMESTEPS}...")

        fpath = os.path.join(xyz_dir, f"t{t}.xyz")
        df_t = load_xyz_data(fpath)
        
        # Interpolate to hex grid
        depth_hex = interpolate_to_hex(df_t, "depth")
        
        # Update max depth
        # Update max depth ignoring NaNs (set first real value, then max)
        valid = ~np.isnan(depth_hex)
        max_depth_hex[valid] = np.where(
            np.isnan(max_depth_hex[valid]),
            depth_hex[valid],
            np.maximum(max_depth_hex[valid], depth_hex[valid])
        )
        
        # Count inundated hex cells and calculate area/volume
        inundated_mask = depth_hex >= DEPTH_THRESHOLD
        # Only count valid hex cells within Shanghai
        valid_mask = np.array([i in valid_hex_ids for i in range(n_hex)])
        inundated_valid = inundated_mask & valid_mask
        
        n_inundated = inundated_valid.sum()
        area_km2 = n_inundated * HEX_AREA_KM2
        
        # Calculate stored water volume
        stored_water_m3 = np.sum(depth_hex[inundated_valid] * HEX_AREA_M2)
        stored_water_million_m3 = stored_water_m3 / 1e6

        # For lowtide (neap tide) scenario, delay time by 8 hours to match the tide data shift
        time_offset = pd.Timedelta(hours=8) if scenario_name == "lowtide" else pd.Timedelta(hours=0)
        current_time = SIM_START + pd.Timedelta(hours=t-1) + time_offset
        area_timeseries.append({
            "timestep": t,
            "datetime": current_time,
            "area_km2": area_km2,
            "n_cells": n_inundated,
            "stored_water_million_m3": stored_water_million_m3
        })
        
        # Update duration (hours above threshold)
        hours_above_hex[inundated_mask] += 1
        if t <= PRE_SURGE_TIMESTEP:
            hours_above_presurge_hex[inundated_mask] += 1

    # Create GeoDataFrames for visualization
    max_depth_gdf = create_hex_gdf(max_depth_hex, "max_depth")
    duration_gdf = create_hex_gdf(hours_above_hex, "hours_above")
    duration_presurge_gdf = create_hex_gdf(hours_above_presurge_hex, "hours_above")
    
    area_ts_df = pd.DataFrame(area_timeseries)

    # Load pre-surge snapshot
    presurge_fpath = os.path.join(xyz_dir, f"t{PRE_SURGE_TIMESTEP}.xyz")
    presurge_df = load_xyz_data(presurge_fpath)
    presurge_depth_hex = interpolate_to_hex(presurge_df, "depth")
    presurge_gdf = create_hex_gdf(presurge_depth_hex, "depth")

    # Compute pre-surge metrics from hex grid
    valid_mask = np.array([i in valid_hex_ids for i in range(n_hex)])
    inundated_presurge = (presurge_depth_hex >= DEPTH_THRESHOLD) & valid_mask
    n_inundated_presurge = inundated_presurge.sum()
    inundated_area_km2 = n_inundated_presurge * HEX_AREA_KM2
    stored_water_m3 = np.sum(presurge_depth_hex[inundated_presurge] * HEX_AREA_M2)
    stored_water_million_m3 = stored_water_m3 / 1e6

    # Store all data
    scenario_data[scenario_name] = {
        "max_depth_gdf": max_depth_gdf,
        "max_depth_hex": max_depth_hex,
        "duration_gdf": duration_gdf,
        "duration_hex": hours_above_hex,
        "duration_presurge_gdf": duration_presurge_gdf,
        "duration_presurge_hex": hours_above_presurge_hex,
        "area_ts_df": area_ts_df,
        "presurge_gdf": presurge_gdf,
        "presurge_depth_hex": presurge_depth_hex,
        "presurge_metrics": {
            "inundated_area_km2": inundated_area_km2,
            "stored_water_million_m3": stored_water_million_m3,
            "n_cells_inundated": n_inundated_presurge
        }
    }

    print(f"  Max depth hex cells: {len(max_depth_gdf)}")
    print(f"  Duration hex cells: {len(duration_gdf)}")
    print(f"  Peak inundated area: {area_ts_df['area_km2'].max():.1f} km²")
    print(f"  Pre-surge inundated area: {inundated_area_km2:.1f} km²")
    print(f"  Pre-surge stored water: {stored_water_million_m3:.1f} million m³")

print("\n" + "="*60)
print("DATA PROCESSING COMPLETE")
print("="*60)
print(f"\nHex grid summary:")
print(f"  Cell area: {HEX_AREA_KM2:.4f} km² ({HEX_AREA_M2:.1f} m²)")
print(f"  Valid cells within Shanghai: {len(valid_hex_ids)}")
print(f"  Total analysis area: {len(valid_hex_ids) * HEX_AREA_KM2:.1f} km²")


```

```python
"""
Wusongkou Station Water Level Extraction from NetCDF Files
===========================================================
Extract water levels at the Wusongkou_1 station from three NetCDF tide model files:

1. FlowFM_0000_his.nc - Astronomical tide only (no surge)
   - Reference time: 1905-08-01 00:00:00
   - Covers: 1905-08-01 to 1905-09-16

2. FlowFM_0000_his_890.nc - Spring tide + surge
   - Reference time: 1905-08-19 00:00:00
   - Covers: 1905-08-19 to 1905-09-02 21:00 (358 hours)
   - Used for: ctrl, norain1, norain2, nostone scenarios

3. FlowFM_0000_his_890_7.nc - Neap tide + surge  
   - Reference time: 1905-08-12 00:00:00
   - Covers: 1905-08-12 to 1905-08-26 21:00 (358 hours)
   - Used for: lowtide (neap) scenario
   - NOTE: Time is shifted to match control run period (1905-08-19 to 1905-09-02)

Station: Wusongkou_1 (station index 12)
"""

import netCDF4 as nc
import numpy as np
import pandas as pd

print("="*60)
print("WUSONGKOU STATION WATER LEVEL EXTRACTION FROM NETCDF")
print("="*60)

# ============================================================================
# PATHS TO NETCDF FILES
# ============================================================================
nc_dir = f"{base_dir}/1905"
nc_astro_path = f"{nc_dir}/FlowFM_0000_his.nc"        # Astronomical tide only
nc_spring_path = f"{nc_dir}/FlowFM_0000_his_890.nc"   # Spring + surge
nc_neap_path = f"{nc_dir}/FlowFM_0000_his_890_7.nc"   # Neap + surge

# Station index for Wusongkou_1
STATION_IDX = 12  # Wusongkou_1 is at index 12

# ============================================================================
# FUNCTION TO EXTRACT WATER LEVEL FROM NETCDF
# ============================================================================
def extract_water_level(nc_path, station_idx, datum_correction=0.0, override_ref_time=None):
    """
    Extract water level time series from NetCDF file at specified station.
    
    Parameters:
    -----------
    nc_path : str
        Path to NetCDF file
    station_idx : int
        Index of station in the NetCDF file
    datum_correction : float
        Datum adjustment to apply (meters)
    override_ref_time : str or None
        If provided, use this as reference time instead of the one in the file
    
    Returns:
    --------
    DataFrame with datetime and water_depth_m columns
    """
    ds = nc.Dataset(nc_path)
    
    # Get time variable
    time_var = ds.variables['time']
    time_units = time_var.units  # e.g., "seconds since 1905-08-19 00:00:00 +00:00"
    
    # Get time values (in seconds since reference)
    time_seconds = time_var[:]
    
    # Determine reference time
    if override_ref_time is not None:
        ref_time = pd.Timestamp(override_ref_time)
    else:
        ref_time_str = time_units.split('since ')[1].split(' +')[0]
        ref_time = pd.Timestamp(ref_time_str)
    
    # Get water level at station
    waterlevel = ds.variables['waterlevel'][:, station_idx]
    
    ds.close()
    
    # Create datetime array
    datetimes = [ref_time + pd.Timedelta(seconds=float(s)) for s in time_seconds]
    
    # Create DataFrame
    df = pd.DataFrame({
        'datetime': datetimes,
        'water_depth_m': waterlevel + datum_correction
    })
    
    return df

# ============================================================================
# EXTRACT WATER LEVELS FROM ALL THREE FILES
# ============================================================================
print("\nExtracting water levels from NetCDF files...")

# 1. Astronomical tide (control, no surge)
print("\n  Loading astronomical tide (no surge)...")
tide_astro = extract_water_level(nc_astro_path, STATION_IDX, datum_correction=2.4)
print(f"    Time range: {tide_astro['datetime'].min()} to {tide_astro['datetime'].max()}")
print(f"    Water level range: {tide_astro['water_depth_m'].min():.2f} to {tide_astro['water_depth_m'].max():.2f} m")
print(f"    Records: {len(tide_astro)}")

# 2. Spring tide + surge (for ctrl, norain1, norain2, nostone)
print("\n  Loading spring tide + surge...")
tide_spring = extract_water_level(nc_spring_path, STATION_IDX, datum_correction=2.4)
print(f"    Time range: {tide_spring['datetime'].min()} to {tide_spring['datetime'].max()}")
print(f"    Water level range: {tide_spring['water_depth_m'].min():.2f} to {tide_spring['water_depth_m'].max():.2f} m")
print(f"    Records: {len(tide_spring)}")

# 3. Neap tide + surge (for lowtide scenario)
# NOTE: Override reference time to match control run period (1905-08-19 00:00:00)
print("\n  Loading neap tide + surge...")
print("    NOTE: Shifting time to match control run period (1905-08-19 start)")
tide_neap = extract_water_level(nc_neap_path, STATION_IDX, datum_correction=2.4, 
                                 override_ref_time="1905-08-19 08:00:00")
print(f"    Time range (shifted): {tide_neap['datetime'].min()} to {tide_neap['datetime'].max()}")
print(f"    Water level range: {tide_neap['water_depth_m'].min():.2f} to {tide_neap['water_depth_m'].max():.2f} m")
print(f"    Records: {len(tide_neap)}")

# ============================================================================
# ASSIGN WATER LEVELS TO SCENARIOS
# ============================================================================
print("\n" + "="*60)
print("ASSIGNING WATER LEVELS TO SCENARIOS")
print("="*60)

# Map scenarios to their appropriate tide data:
# - ctrl, norain1, norain2, nostone: spring tide + surge
# - lowtide: neap tide + surge (time shifted to match control)
scenario_tide_mapping = {
    "ctrl": "spring",
    "lowtide": "neap",
    "norain1": "spring",
    "norain2": "spring",
    "nostone": "spring"
}

wusong_simulated = {}

for scenario_name in scenario_dirs.keys():
    tide_type = scenario_tide_mapping.get(scenario_name, "spring")
    
    if tide_type == "spring":
        wusong_df = tide_spring.copy()
    else:  # neap
        wusong_df = tide_neap.copy()
    
    wusong_df['timestep'] = range(1, len(wusong_df) + 1)
    wusong_simulated[scenario_name] = wusong_df
    
    print(f"\n  {scenario_name}: using {tide_type} tide + surge")
    print(f"    Time range: {wusong_df['datetime'].min()} to {wusong_df['datetime'].max()}")
    print(f"    Water level range: {wusong_df['water_depth_m'].min():.2f} to {wusong_df['water_depth_m'].max():.2f} m")

# ============================================================================
# SAVE TO CSV FILES
# ============================================================================
print("\n" + "="*60)
print("SAVING TO CSV FILES")
print("="*60)

for scenario_name, wusong_df in wusong_simulated.items():
    output_path = f"{base_dir}/wusongkou_simulated_{scenario_name}.csv"
    wusong_df.to_csv(output_path, index=False)
    print(f"  Saved: {output_path}")

print("\n" + "="*60)
print("EXTRACTION COMPLETE")
print("="*60)
print("\nScenario-tide mapping:")
print("  ctrl, norain1, norain2, nostone -> Spring tide + surge")
print("  lowtide -> Neap tide + surge (time shifted to 1905-08-19 start)")
print("\nAll scenarios now have the same time range: 1905-08-19 to 1905-09-02")
print("\nAstronomical tide (no surge) available in tide_astro DataFrame")
```

```python
"""
Figure 4: Tide-locking as a Drainage Failure Mode (Nature Cities)
=================================================================
- Fig. 4A: Water-level time series showing spring vs neap tide relative to interior stages
- Fig. 4B: Control vs Neap-Tide (lowtide) inundation area time series and recession comparison
- Fig. 4C: Neap-Tide minus Control duration difference map

Uses pre-processed data from data processing cell.
Uses simulated water level from NetCDF files:
  - Spring+surge (ctrl): 1905-08-19 to 1905-09-02
  - Neap+surge (lowtide): 1905-08-12 to 1905-08-26

NOTE: The neap tide scenario covers a different time period than the control.
Panel 4A shows overlapping period (08-20 to 08-26) for direct comparison.
"""

# Interior outfall stage (representative threshold for tide-locking)
INTERIOR_OUTFALL_STAGE = 3.0  # meters

# ============================================================================
# COMPUTE TIDE-LOCKING INDEX
# ============================================================================
print("Computing Tide-Locking Index...")

# Load simulated water levels for control and lowtide (neap)
wusong_ctrl = wusong_simulated["ctrl"].copy()
wusong_neap = wusong_simulated["lowtide"].copy()

print(f"Control (spring+surge) time range: {wusong_ctrl['datetime'].min()} to {wusong_ctrl['datetime'].max()}")
print(f"Neap-Tide (neap+surge) time range: {wusong_neap['datetime'].min()} to {wusong_neap['datetime'].max()}")

# Hours above outfall stage
hours_above_outfall_ctrl = (wusong_ctrl["water_depth_m"] >= INTERIOR_OUTFALL_STAGE).sum()
hours_above_outfall_neap = (wusong_neap["water_depth_m"] >= INTERIOR_OUTFALL_STAGE).sum()

tide_lock_fraction_ctrl = hours_above_outfall_ctrl / len(wusong_ctrl) if len(wusong_ctrl) > 0 else 0
tide_lock_fraction_neap = hours_above_outfall_neap / len(wusong_neap) if len(wusong_neap) > 0 else 0

print(f"Tide-locking fraction (Control): {tide_lock_fraction_ctrl:.2%} ({hours_above_outfall_ctrl} hours)")
print(f"Tide-locking fraction (Neap-Tide): {tide_lock_fraction_neap:.2%} ({hours_above_outfall_neap} hours)")

# Head difference integral
head_diff_ctrl = np.maximum(0, wusong_ctrl["water_depth_m"].values - INTERIOR_OUTFALL_STAGE)
head_diff_neap = np.maximum(0, wusong_neap["water_depth_m"].values - INTERIOR_OUTFALL_STAGE)

integral_ctrl = np.sum(head_diff_ctrl)  # meter-hours
integral_neap = np.sum(head_diff_neap)

print(f"Head difference integral (Control): {integral_ctrl:.1f} m·h")
print(f"Head difference integral (Neap-Tide): {integral_neap:.1f} m·h")

# ============================================================================
# COMPUTE DIFFERENCE MAP (Neap-Tide minus Control)
# ============================================================================
print("\nComputing duration difference map...")

ctrl_dur = scenario_data["ctrl"]["duration_gdf"].copy()
neap_dur = scenario_data["lowtide"]["duration_gdf"].copy()

# Use hex_id for merging
dur_merged = neap_dur.merge(
    ctrl_dur[["hex_id", "hours_above"]],
    on="hex_id",
    how="outer",
    suffixes=("_neap", "_ctrl")
)
dur_merged["hours_above_neap"] = dur_merged["hours_above_neap"].fillna(0)
dur_merged["hours_above_ctrl"] = dur_merged["hours_above_ctrl"].fillna(0)
dur_merged["duration_diff"] = dur_merged["hours_above_neap"] - dur_merged["hours_above_ctrl"]

dur_merged_gdf = gpd.GeoDataFrame(dur_merged, geometry="geometry", crs=neap_dur.crs)

print(f"Duration difference range: {dur_merged['duration_diff'].min():.0f} to {dur_merged['duration_diff'].max():.0f} hours")

# ============================================================================
# RECESSION ANALYSIS
# ============================================================================
print("\nAnalyzing recession curves...")

ctrl_ts = scenario_data["ctrl"]["area_ts_df"]
neap_ts = scenario_data["lowtide"]["area_ts_df"]

ctrl_peak_idx = ctrl_ts["area_km2"].idxmax()
neap_peak_idx = neap_ts["area_km2"].idxmax()

ctrl_peak_time = ctrl_ts.loc[ctrl_peak_idx, "datetime"]
neap_peak_time = neap_ts.loc[neap_peak_idx, "datetime"]
ctrl_peak_area = ctrl_ts.loc[ctrl_peak_idx, "area_km2"]
neap_peak_area = neap_ts.loc[neap_peak_idx, "area_km2"]

print(f"Control peak: {ctrl_peak_area:.1f} km² at {ctrl_peak_time}")
print(f"Neap-Tide peak: {neap_peak_area:.1f} km² at {neap_peak_time}")

# ============================================================================
# FIGURE 4: THREE-PANEL COMPOSITE
# ============================================================================
fig = plt.figure(figsize=(12, 10), dpi=150)

gs = gridspec.GridSpec(2, 2, height_ratios=[0.8, 1.2], width_ratios=[1, 1],
                       hspace=0.30, wspace=0.25)

ax_tide = fig.add_subplot(gs[0, 0])      # 4A: tide time series
ax_recession = fig.add_subplot(gs[0, 1]) # 4B: recession curves
ax_map = fig.add_subplot(gs[1, :])       # 4C: difference map (full width)

# ============================================================================
# PANEL 4A: WATER LEVEL TIME SERIES
# ============================================================================
# Note: Control (spring+surge) covers 08-19 to 09-02
#       Neap (neap+surge) covers 08-12 to 08-26
# We show the full time range of each, overlapping where they both exist

# Plot control (spring+surge) - full range
ax_tide.plot(wusong_ctrl["datetime"], wusong_ctrl["water_depth_m"],
             color="#d62728", linewidth=1, label="Spring Tide + Surge (Control)")

# Plot neap+surge - full range
ax_tide.plot(wusong_neap["datetime"], wusong_neap["water_depth_m"],
             color="#1f77b4", linewidth=1, label="Neap Tide + Surge")

# Interior outfall threshold
ax_tide.axhline(INTERIOR_OUTFALL_STAGE, color="green", linestyle="--", linewidth=1.5,
                label=f"Outfall Stage ({INTERIOR_OUTFALL_STAGE} m)")

# Shade tide-locked periods: fill area between curve and outfall stage (when above)
# Control (spring+surge) - red fill
ctrl_water_levels = wusong_ctrl["water_depth_m"].values
ctrl_times = wusong_ctrl["datetime"].values
ax_tide.fill_between(ctrl_times, INTERIOR_OUTFALL_STAGE, ctrl_water_levels,
                     where=(ctrl_water_levels >= INTERIOR_OUTFALL_STAGE),
                     alpha=0.3, color="#d62728", label="_nolegend_")

# Neap tide - blue fill
neap_water_levels = wusong_neap["water_depth_m"].values
neap_times = wusong_neap["datetime"].values
ax_tide.fill_between(neap_times, INTERIOR_OUTFALL_STAGE, neap_water_levels,
                     where=(neap_water_levels >= INTERIOR_OUTFALL_STAGE),
                     alpha=0.3, color="#1f77b4", label="_nolegend_")

ax_tide.set_xlabel("Date (1905)", fontsize=10)
ax_tide.set_ylabel("Water Level at Wusongkou (m)", fontsize=10)

# Set x limits to show full range of both datasets
all_times = list(wusong_ctrl["datetime"]) + list(wusong_neap["datetime"])
ax_tide.set_xlim(min(all_times), max(all_times))
ax_tide.set_ylim(0, max(wusong_ctrl["water_depth_m"].max(), wusong_neap["water_depth_m"].max()) * 1.1)

ax_tide.xaxis.set_major_locator(mdates.DayLocator(interval=2))
ax_tide.xaxis.set_major_formatter(mdates.DateFormatter('%m-%d'))
ax_tide.tick_params(axis='x', rotation=45)

ax_tide.legend(loc='upper left', fontsize=7, frameon=True)
ax_tide.grid(True, alpha=0.3)

# Add tide-locking index annotation
tide_lock_text = (f"Tide-Lock Index:\n"
                  f"Spring+Surge: {tide_lock_fraction_ctrl:.0%}\n"
                  f"Neap+Surge: {tide_lock_fraction_neap:.0%}")
ax_tide.text(0.98, 0.98, tide_lock_text, transform=ax_tide.transAxes, fontsize=8,
             verticalalignment='top', horizontalalignment='right',
             bbox=dict(boxstyle='round', facecolor='white', alpha=0.9))

ax_tide.set_title("A. Water level at Wusongkou: Spring+Surge vs Neap+Surge",
                  fontsize=10, fontweight="bold", loc="left")

# ============================================================================
# PANEL 4B: INUNDATION AREA AND RECESSION CURVES
# ============================================================================
ax_recession.plot(ctrl_ts["datetime"], ctrl_ts["area_km2"],
                  color="#d62728", linewidth=1.5, label="Control (Spring Tide)")
ax_recession.plot(neap_ts["datetime"], neap_ts["area_km2"],
                  color="#1f77b4", linewidth=1.5, label="Neap-Tide Scenario")

# Mark peaks
ax_recession.scatter([ctrl_peak_time], [ctrl_peak_area], color="#d62728", s=80,
                     zorder=5, edgecolor='black', marker='o')
ax_recession.scatter([neap_peak_time], [neap_peak_area], color="#1f77b4", s=80,
                     zorder=5, edgecolor='black', marker='o')

# Mark typhoon arrival (only relevant for control)
typhoon_time = pd.Timestamp("1905-09-01 12:00")
if typhoon_time <= ctrl_ts["datetime"].max():
    ax_recession.axvline(typhoon_time, color="purple", linestyle=":", linewidth=1.5)
    ax_recession.text(typhoon_time, ax_recession.get_ylim()[1]*0.95, "Typhoon\nSurge",
                      color="purple", fontsize=8, ha='center', va='top')

ax_recession.set_xlabel("Date (1905)", fontsize=10)
ax_recession.set_ylabel(f"Inundated Area > {DEPTH_THRESHOLD} m (km²)", fontsize=10)
ax_recession.xaxis.set_major_locator(mdates.DayLocator(interval=2))
ax_recession.xaxis.set_major_formatter(mdates.DateFormatter('%m-%d'))
ax_recession.tick_params(axis='x', rotation=45)

ax_recession.legend(loc='upper left', fontsize=8, frameon=True)
ax_recession.grid(True, alpha=0.3)

# Summary statistics
diff_peak = ctrl_peak_area - neap_peak_area
stats_text = (f"Peak difference: {diff_peak:.0f} km²\n"
              f"Control peak: {ctrl_peak_area:.0f} km²\n"
              f"Neap peak: {neap_peak_area:.0f} km²")
ax_recession.text(0.98, 0.60, stats_text, transform=ax_recession.transAxes, fontsize=8,
                  verticalalignment='top', horizontalalignment='right',
                  bbox=dict(boxstyle='round', facecolor='white', alpha=0.9))

ax_recession.set_title("B. Inundated area time series: Control vs Neap-Tide",
                       fontsize=10, fontweight="bold", loc="left")

# ============================================================================
# PANEL 4C: DURATION DIFFERENCE MAP
# ============================================================================
setup_map_ax(ax_map)

# Treat 0 as NaN for plotting (don't show cells with no difference)
dur_merged_gdf_plot = dur_merged_gdf.copy()
dur_merged_gdf_plot.loc[dur_merged_gdf_plot["duration_diff"] == 0, "duration_diff"] = np.nan

max_abs_dur = max(abs(dur_merged["duration_diff"].min()), abs(dur_merged["duration_diff"].max()))
if max_abs_dur == 0:
    max_abs_dur = 1
norm_dur = mcolors.TwoSlopeNorm(vmin=-max_abs_dur/2., vcenter=0, vmax=max_abs_dur/2.)

hb = plot_hexbin_map(ax_map, dur_merged_gdf_plot, "duration_diff", CMAP_DIFF, norm_dur)

cbar = plt.colorbar(hb, ax=ax_map, orientation="vertical", shrink=0.7, pad=0.02, extend='both')
cbar.set_label(f"Duration difference (hours, depth > {DEPTH_THRESHOLD} m)", fontsize=9)

ax_map.legend(handles=LEGEND_ELEMENTS, loc='lower center', ncol=5, fontsize=7,
              frameon=True, framealpha=0.9, bbox_to_anchor=(0.5, -0.02))

ax_map.set_title("C. Duration difference map (Neap-Tide − Control): Red = longer flooding under neap tide",
                 fontsize=10, fontweight="bold", loc="left")

# ============================================================================
# FINAL ADJUSTMENTS
# ============================================================================
plt.suptitle("Figure 4: Tide-locking as a Drainage Failure Mode",
             fontsize=12, fontweight="bold", y=0.98)
plt.tight_layout(rect=[0, 0, 1, 0.96])

fig.savefig(f"{base_dir}/Fig4_TideLocking_NCities.pdf", dpi=300, bbox_inches="tight")
fig.savefig(f"{base_dir}/Fig4_TideLocking_NCities.png", dpi=300, bbox_inches="tight")

plt.show()

# ============================================================================
# SUMMARY
# ============================================================================
print("\n" + "="*60)
print("SUMMARY: TIDE-LOCKING ANALYSIS")
print("="*60)
print(f"\nTide-Locking Index (hours above {INTERIOR_OUTFALL_STAGE}m outfall):")
print(f"  Control (Spring+Surge): {tide_lock_fraction_ctrl:.1%} ({hours_above_outfall_ctrl} hours)")
print(f"  Neap+Surge: {tide_lock_fraction_neap:.1%} ({hours_above_outfall_neap} hours)")
print(f"\nHead difference integral:")
print(f"  Control: {integral_ctrl:.1f} m·h")
print(f"  Neap-Tide: {integral_neap:.1f} m·h")
print(f"\nPeak inundation:")
print(f"  Control: {ctrl_peak_area:.0f} km² at {ctrl_peak_time}")
print(f"  Neap-Tide: {neap_peak_area:.0f} km² at {neap_peak_time}")
print(f"  Difference: {diff_peak:.0f} km²")

print("\nFigure 4 saved to:")
print(f"  - {base_dir}/Fig4_TideLocking_NCities.pdf")
print(f"  - {base_dir}/Fig4_TideLocking_NCities.png")
```

```python
"""
Figure 5: Defenses Redistribute Inundation Rather Than Simply Reduce It
========================================================================
Purpose: Analyze how stone embankments redistribute flood impacts spatially.

Core comparison: No-Stone (earthen embankments only) vs Control (with stone embankments)

Analyses:
- Redistribution decomposition: quantify "winners vs losers" hydraulically
  - % area with depth increase vs decrease
  - Total positive/negative depth-volume change (depth diff × cell area)
  - Gini coefficient on absolute depth change for spatial concentration
- Hotspot shift: identify where largest increases occur (top decile)

Figures:
- Fig. 5A: Max depth difference map (No-Stone minus Control)
- Fig. 5B: Summary bars of positive/negative changes and area shares
"""

# ============================================================================
# COMPUTE DIFFERENCE MAPS (No-Stone minus Control)
# ============================================================================
print("Computing difference maps (No-Stone minus Control)...")
print("No-Stone scenario = earthen embankments only (no stone seawalls)")
print("Control scenario = with stone embankments")

# Get max depth GeoDataFrames
ctrl_maxdepth = scenario_data["ctrl"]["max_depth_gdf"].copy()
nostone_maxdepth = scenario_data["nostone"]["max_depth_gdf"].copy()

# Get duration GeoDataFrames for later use
ctrl_dur = scenario_data["ctrl"]["duration_gdf"].copy()
nostone_dur = scenario_data["nostone"]["duration_gdf"].copy()

# Create coordinate keys for merging
# Merge max depth data using hex_id
depth_merged = nostone_maxdepth.merge(
    ctrl_maxdepth[["hex_id", "max_depth", "geometry"]],
    on="hex_id",
    how="outer",
    suffixes=("_nostone", "_ctrl")
)
# If geometry missing (hex only in ctrl), pull it from ctrl
if "geometry_ctrl" in depth_merged.columns:
    # geometry was renamed to geometry_nostone due to merge suffixes
    depth_merged["geometry"] = depth_merged["geometry_nostone"].fillna(depth_merged["geometry_ctrl"])
    depth_merged = depth_merged.drop(columns=["geometry_nostone", "geometry_ctrl"])

depth_merged["max_depth_nostone"] = depth_merged["max_depth_nostone"].fillna(0)
depth_merged["max_depth_ctrl"] = depth_merged["max_depth_ctrl"].fillna(0)
depth_merged["depth_diff"] = depth_merged["max_depth_nostone"] - depth_merged["max_depth_ctrl"]

depth_merged_gdf = gpd.GeoDataFrame(depth_merged, geometry="geometry", crs=nostone_maxdepth.crs)

# Merge duration data using hex_id
dur_merged = nostone_dur.merge(
    ctrl_dur[["hex_id", "hours_above", "geometry"]],
    on="hex_id",
    how="outer",
    suffixes=("_nostone", "_ctrl")
)
# If geometry missing (hex only in ctrl), pull it from ctrl
if "geometry_ctrl" in dur_merged.columns:
    # geometry was renamed to geometry_nostone due to merge suffixes
    dur_merged["geometry"] = dur_merged["geometry_nostone"].fillna(dur_merged["geometry_ctrl"])
    dur_merged = dur_merged.drop(columns=["geometry_nostone", "geometry_ctrl"])

dur_merged["hours_above_nostone"] = dur_merged["hours_above_nostone"].fillna(0)
dur_merged["hours_above_ctrl"] = dur_merged["hours_above_ctrl"].fillna(0)
dur_merged["duration_diff"] = dur_merged["hours_above_nostone"] - dur_merged["hours_above_ctrl"]

dur_merged_gdf = gpd.GeoDataFrame(dur_merged, geometry="geometry", crs=nostone_dur.crs)

# ============================================================================
# REDISTRIBUTION DECOMPOSITION ANALYSIS
# ============================================================================
print("\n" + "="*60)
print("REDISTRIBUTION DECOMPOSITION ANALYSIS")
print("="*60)

# Filter to cells with any flooding in either scenario
flooded_cells = depth_merged[
    (depth_merged["max_depth_nostone"] > 0.01) | (depth_merged["max_depth_ctrl"] > 0.01)
].copy()

n_total_cells = len(flooded_cells)
cell_area_m2 = HEX_AREA_KM2 * 1e6

# Classify cells
flooded_cells["change_type"] = "no_change"
flooded_cells.loc[flooded_cells["depth_diff"] > 0.01, "change_type"] = "increase"  # Worse without stone
flooded_cells.loc[flooded_cells["depth_diff"] < -0.01, "change_type"] = "decrease"  # Better without stone

# Count cells by change type
n_increase = (flooded_cells["change_type"] == "increase").sum()
n_decrease = (flooded_cells["change_type"] == "decrease").sum()
n_no_change = (flooded_cells["change_type"] == "no_change").sum()

pct_increase = n_increase / n_total_cells * 100
pct_decrease = n_decrease / n_total_cells * 100
pct_no_change = n_no_change / n_total_cells * 100

print(f"\nCell classification (flooded in either scenario):")
print(f"  Total flooded cells: {n_total_cells}")
print(f"  Depth INCREASE (worse without stone): {n_increase} ({pct_increase:.1f}%)")
print(f"  Depth DECREASE (better without stone): {n_decrease} ({pct_decrease:.1f}%)")
print(f"  No significant change: {n_no_change} ({pct_no_change:.1f}%)")

# Compute depth-volume changes
# Positive depth-volume change = areas that got deeper without stone embankments
positive_depth_change = flooded_cells[flooded_cells["depth_diff"] > 0]["depth_diff"].values
negative_depth_change = flooded_cells[flooded_cells["depth_diff"] < 0]["depth_diff"].values

total_positive_volume_m3 = np.sum(positive_depth_change) * cell_area_m2
total_negative_volume_m3 = np.sum(np.abs(negative_depth_change)) * cell_area_m2

total_positive_volume_million_m3 = total_positive_volume_m3 / 1e6
total_negative_volume_million_m3 = total_negative_volume_m3 / 1e6

print(f"\nDepth-Volume Changes:")
print(f"  Total POSITIVE depth-volume (deeper without stone): {total_positive_volume_million_m3:.2f} million m³")
print(f"  Total NEGATIVE depth-volume (shallower without stone): {total_negative_volume_million_m3:.2f} million m³")
print(f"  Net change: {(total_positive_volume_million_m3 - total_negative_volume_million_m3):.2f} million m³")

# Area calculations
area_increase_km2 = n_increase * HEX_AREA_KM2
area_decrease_km2 = n_decrease * HEX_AREA_KM2
area_no_change_km2 = n_no_change * HEX_AREA_KM2

print(f"\nArea by change type:")
print(f"  Area with depth INCREASE: {area_increase_km2:.1f} km²")
print(f"  Area with depth DECREASE: {area_decrease_km2:.1f} km²")
print(f"  Area with no change: {area_no_change_km2:.1f} km²")

# ============================================================================
# GINI COEFFICIENT FOR SPATIAL CONCENTRATION
# ============================================================================
def gini_coefficient(values):
    """Calculate Gini coefficient for a set of values."""
    values = np.array(values)
    values = values[~np.isnan(values)]
    if len(values) == 0:
        return 0
    values = np.sort(values)
    n = len(values)
    cumulative = np.cumsum(values)
    return (2 * np.sum((np.arange(1, n+1) * values)) - (n + 1) * cumulative[-1]) / (n * cumulative[-1]) if cumulative[-1] > 0 else 0

abs_depth_change = np.abs(flooded_cells["depth_diff"].values)
gini = gini_coefficient(abs_depth_change)

print(f"\nSpatial Concentration:")
print(f"  Gini coefficient of |depth change|: {gini:.3f}")
print(f"  (0 = uniform change everywhere, 1 = all change in one cell)")

# ============================================================================
# HOTSPOT ANALYSIS: TOP DECILE OF DEPTH INCREASES
# ============================================================================
print("\n" + "="*60)
print("HOTSPOT ANALYSIS: TOP DECILE OF DEPTH INCREASES")
print("="*60)

# Get cells with depth increases
increase_cells = flooded_cells[flooded_cells["depth_diff"] > 0.01].copy()

if len(increase_cells) > 0:
    # Find top 10% of depth increases
    threshold_90 = np.percentile(increase_cells["depth_diff"], 90)
    hotspots = increase_cells[increase_cells["depth_diff"] >= threshold_90]
    
    n_hotspots = len(hotspots)
    hotspot_area_km2 = n_hotspots * HEX_AREA_KM2
    mean_increase_hotspot = hotspots["depth_diff"].mean()
    max_increase_hotspot = hotspots["depth_diff"].max()
    
    print(f"Top decile threshold: {threshold_90:.2f} m increase")
    print(f"Number of hotspot cells: {n_hotspots}")
    print(f"Hotspot area: {hotspot_area_km2:.1f} km²")
    print(f"Mean depth increase in hotspots: {mean_increase_hotspot:.2f} m")
    print(f"Maximum depth increase: {max_increase_hotspot:.2f} m")
else:
    hotspots = None
    threshold_90 = 0
    print("No cells with depth increase found.")

# ============================================================================
# FIGURE 5: TWO-PANEL COMPOSITE
# ============================================================================
print("\n" + "="*60)
print("GENERATING FIGURE 5")
print("="*60)

fig = plt.figure(figsize=(14, 6), dpi=150)

gs = gridspec.GridSpec(1, 2, width_ratios=[1.2, 0.8], wspace=0.25)

ax_map = fig.add_subplot(gs[0])     # 5A: difference map
ax_bars = fig.add_subplot(gs[1])    # 5B: summary bars

# ============================================================================
# PANEL 5A: MAX DEPTH DIFFERENCE MAP (No-Stone minus Control)
# ============================================================================
print("Plotting Panel 5A: Max depth difference map...")

setup_map_ax(ax_map)

# Diverging colormap centered at 0
max_abs = max(abs(depth_merged["depth_diff"].min()), abs(depth_merged["depth_diff"].max()))
if max_abs == 0:
    max_abs = 0.1
# Clip to reasonable range for visualization
max_abs = min(max_abs, 1.0)
norm_diff = mcolors.TwoSlopeNorm(vmin=-max_abs, vcenter=0, vmax=max_abs)

hb = plot_hexbin_map(ax_map, depth_merged_gdf, "depth_diff", CMAP_DIFF, norm_diff)

# Add hotspot markers if available
if hotspots is not None and len(hotspots) > 0:
    # Use centroid of hex polygons to get point coordinates
    hotspot_gdf = gpd.GeoDataFrame(
        hotspots,
        geometry=gpd.points_from_xy(
            [depth_merged_gdf.loc[depth_merged_gdf["hex_id"] == ck, "geometry"].values[0].centroid.x 
             if len(depth_merged_gdf.loc[depth_merged_gdf["hex_id"] == ck]) > 0 else np.nan 
             for ck in hotspots["hex_id"]],
            [depth_merged_gdf.loc[depth_merged_gdf["hex_id"] == ck, "geometry"].values[0].centroid.y 
             if len(depth_merged_gdf.loc[depth_merged_gdf["hex_id"] == ck]) > 0 else np.nan 
             for ck in hotspots["hex_id"]]
        ),
        crs=depth_merged_gdf.crs
    )

cbar = plt.colorbar(hb, ax=ax_map, orientation="vertical", shrink=0.7, pad=0.02, extend='both')
cbar.set_label("Max depth difference (m)\n[No-Stone − Control]", fontsize=9)

# Custom legend
legend_elements_map = LEGEND_ELEMENTS.copy()
ax_map.legend(handles=legend_elements_map, loc='lower center', ncol=3, fontsize=7,
              frameon=True, framealpha=0.9, bbox_to_anchor=(0.5, -0.02))

ax_map.set_title("A. Max depth difference (No-Stone − Control)\nRed = deeper without stone embankments, Blue = shallower",
                 fontsize=10, fontweight="bold", loc="left")

# ============================================================================
# PANEL 5B: SUMMARY BARS
# ============================================================================
print("Plotting Panel 5B: Summary statistics...")

# Create subplots within ax_bars area
# We'll use the axes directly for grouped bars

# Data for plotting
categories = ['Area\nAffected', 'Depth-Volume\nChange']

# For area: increase vs decrease
area_data = [area_increase_km2, area_decrease_km2]

# For volume: positive vs negative (in absolute terms for comparison)
volume_data = [total_positive_volume_million_m3, total_negative_volume_million_m3]

x = np.arange(2)
width = 0.35

# Create twin axis for volume
ax_bars2 = ax_bars.twinx()

# Plot bars
bars1 = ax_bars.bar(0 - width/2, area_increase_km2, width, label='Depth Increase (worse)', 
                    color='#d62728', alpha=0.8)
bars2 = ax_bars.bar(0 + width/2, area_decrease_km2, width, label='Depth Decrease (better)', 
                    color='#1f77b4', alpha=0.8)

bars3 = ax_bars2.bar(1 - width/2, total_positive_volume_million_m3, width, 
                     color='#d62728', alpha=0.8, hatch='//')
bars4 = ax_bars2.bar(1 + width/2, total_negative_volume_million_m3, width, 
                     color='#1f77b4', alpha=0.8, hatch='//')

# Labels
ax_bars.set_ylabel('Area (km²)', fontsize=10)
ax_bars2.set_ylabel('Volume change (million m³)', fontsize=10)

ax_bars.set_xticks([0, 1])
ax_bars.set_xticklabels(['Area\nAffected', 'Depth-Volume\nChange'], fontsize=10)

# Add value labels
ax_bars.text(0 - width/2, area_increase_km2 + 5, f'{area_increase_km2:.0f}\n({pct_increase:.0f}%)', 
             ha='center', va='bottom', fontsize=8, color='#d62728')
ax_bars.text(0 + width/2, area_decrease_km2 + 5, f'{area_decrease_km2:.0f}\n({pct_decrease:.0f}%)', 
             ha='center', va='bottom', fontsize=8, color='#1f77b4')

ax_bars2.text(1 - width/2, total_positive_volume_million_m3 + 0.5, f'{total_positive_volume_million_m3:.1f}', 
              ha='center', va='bottom', fontsize=8, color='#d62728')
ax_bars2.text(1 + width/2, total_negative_volume_million_m3 + 0.5, f'{total_negative_volume_million_m3:.1f}', 
              ha='center', va='bottom', fontsize=8, color='#1f77b4')

# Legend
from matplotlib.patches import Patch
legend_elements_bar = [
    Patch(facecolor='#d62728', alpha=0.8, label='Depth increase (worse without stone)'),
    Patch(facecolor='#1f77b4', alpha=0.8, label='Depth decrease (better without stone)'),
]
ax_bars.legend(handles=legend_elements_bar, loc='upper right', fontsize=8, frameon=True)

# Add summary text box
summary_text = (
    f"Redistribution Summary\n"
    f"─────────────────\n"
    f"Gini coefficient: {gini:.3f}\n"
    f"(spatial concentration)\n\n"
    f"Net volume change:\n"
    f"{(total_positive_volume_million_m3 - total_negative_volume_million_m3):.2f} million m³\n\n"
    f"Hotspots (top 10%):\n"
    f"  Area: {hotspot_area_km2:.1f} km²\n"
    f"  Mean Δdepth: +{mean_increase_hotspot:.2f} m" if hotspots is not None and len(hotspots) > 0 else ""
)
ax_bars.text(0.02, 0.98, summary_text, transform=ax_bars.transAxes, fontsize=8,
             verticalalignment='top', horizontalalignment='left',
             bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.9),
             family='monospace')

ax_bars.set_title("B. Redistribution decomposition:\nWinners vs Losers",
                  fontsize=10, fontweight="bold", loc="left")

# ============================================================================
# FINAL ADJUSTMENTS
# ============================================================================
plt.suptitle("Figure 5: Stone Embankments Redistribute Rather Than Simply Reduce Inundation",
             fontsize=12, fontweight="bold", y=0.98)
plt.tight_layout(rect=[0, 0, 1, 0.95])

fig.savefig(f"{base_dir}/Fig5_Redistribution_NCities.pdf", dpi=300, bbox_inches="tight")
fig.savefig(f"{base_dir}/Fig5_Redistribution_NCities.png", dpi=300, bbox_inches="tight")

plt.show()

# ============================================================================
# SUMMARY
# ============================================================================
print("\n" + "="*60)
print("SUMMARY: EMBANKMENT REDISTRIBUTION ANALYSIS")
print("="*60)
print(f"\nComparison: No-Stone (earthen only) vs Control (with stone)")
print(f"\nSpatial Redistribution:")
print(f"  Cells with depth INCREASE: {n_increase} ({pct_increase:.1f}%) = {area_increase_km2:.1f} km²")
print(f"  Cells with depth DECREASE: {n_decrease} ({pct_decrease:.1f}%) = {area_decrease_km2:.1f} km²")
print(f"  Cells with no change: {n_no_change} ({pct_no_change:.1f}%) = {area_no_change_km2:.1f} km²")
print(f"\nVolume Redistribution:")
print(f"  Positive depth-volume change: +{total_positive_volume_million_m3:.2f} million m³")
print(f"  Negative depth-volume change: -{total_negative_volume_million_m3:.2f} million m³")
print(f"  Net change: {(total_positive_volume_million_m3 - total_negative_volume_million_m3):.2f} million m³")
print(f"\nSpatial Concentration:")
print(f"  Gini coefficient: {gini:.3f}")
if hotspots is not None and len(hotspots) > 0:
    print(f"\nHotspots (top 10% of depth increases):")
    print(f"  Threshold: >{threshold_90:.2f} m increase")
    print(f"  Number of cells: {n_hotspots}")
    print(f"  Area: {hotspot_area_km2:.1f} km²")
    print(f"  Mean increase: {mean_increase_hotspot:.2f} m")
    print(f"  Max increase: {max_increase_hotspot:.2f} m")

print("\nFigure 5 saved to:")
print(f"  - {base_dir}/Fig5_Redistribution_NCities.pdf")
print(f"  - {base_dir}/Fig5_Redistribution_NCities.png")
```

```python
"""
Figure 6: A Compact Diagnostic Summary Across Experiments
=========================================================
Purpose: Provide a "portable" city diagnostic across all scenarios.

Metrics computed for each run:
1. Max inundated area (km²)
2. Area > 1m depth (km²)
3. Hours-area integral above 0.3m (persistence-disruption proxy)
4. Pre-surge stored water proxy (million m³)
5. Tide-locking Index (peak area timing relative to high tide)
6. Redistribution metric (only for nostone experiment vs ctrl)

Figures:
- Fig 6A: Radar plot comparing scenarios
- Fig 6B: Stacked bars (simple, interpretable)
- Fig 6C: Diagnostic space scatter (Preconditioning vs Tide-locking, marker size = persistence)
"""

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from matplotlib.patches import Patch
from matplotlib.lines import Line2D

# ============================================================================
# COMPUTE DIAGNOSTIC METRICS FOR ALL SCENARIOS
# ============================================================================
print("="*60)
print("COMPUTING DIAGNOSTIC METRICS FOR ALL SCENARIOS")
print("="*60)

# Define scenarios to analyze (exclude nostone for now, add separately)
scenarios_to_analyze = ["ctrl", "lowtide", "norain1", "norain2", "nostone"]

# Dictionary to store metrics
diagnostic_metrics = {}

for scenario in scenarios_to_analyze:
    if scenario not in scenario_data:
        print(f"Skipping {scenario} - not in scenario_data")
        continue
    
    data = scenario_data[scenario]
    area_ts = data["area_ts_df"]
    max_depth_gdf = data["max_depth_gdf"]
    presurge_metrics = data["presurge_metrics"]
    
    print(f"\n--- {scenario.upper()} ---")
    
    # 1. Max inundated area (km²)
    max_inundated_area = area_ts["area_km2"].max()
    
    # 2. Area > 1m depth (km²)
    area_above_1m = (max_depth_gdf["max_depth"] > 1.0).sum() * hex_grid_info['area_km2']
    
    # 3. Hours-area integral above 0.3m (persistence-disruption proxy)
    # This is the integral of (area above 0.3m) over time
    # We need to compute area above 0.3m for each timestep
    hours_area_integral = 0
    for t in range(1, 359):
        # Get depth at this timestep from the stored data
        # Since we have area_ts which tracks total flooded area,
        # we approximate by using the area_km2 weighted by typical depth ratio
        # Better: use actual depth threshold from area_ts if available
        pass
    
    # Approximate using area_ts - assume most flooded area is > 0.3m during peak
    # More accurate: compute from raw data, but area_ts gives good proxy
    # Hours-area integral = sum of (area * dt) where dt = 1 hour
    hours_area_integral = area_ts["area_km2"].sum()  # km² × hours
    
    # 4. Pre-surge stored water proxy (million m³)
    presurge_stored_water = presurge_metrics["stored_water_million_m3"]
    
    # 5. Tide-locking Index
    # Compute as: timing of peak inundation relative to high tide
    # High tide timing from astronomical tide data
    peak_idx = area_ts["area_km2"].idxmax()
    peak_time = area_ts.loc[peak_idx, "datetime"]
    
    # Find nearest high tide from astronomical tide data
    if "tide_astro" in dir() or "tide_astro" in globals():
        # Find high tide times (local maxima in tide level)
        tide_levels = tide_astro["water_depth_m"].values
        tide_times = pd.to_datetime(tide_astro["datetime"])
        
        # Simple: find the tide level at peak inundation time
        tide_at_peak_idx = np.argmin(np.abs(tide_times - peak_time))
        tide_at_peak = tide_levels[tide_at_peak_idx]
        max_tide = tide_levels.max()
        min_tide = tide_levels.min()
        
        # Tide-locking index: 1 = peak at high tide, 0 = peak at low tide
        tide_locking_index = (tide_at_peak - min_tide) / (max_tide - min_tide) if (max_tide - min_tide) > 0 else 0.5
    else:
        # Fallback if tide_astro not available
        tide_locking_index = 0.5
    
    # 6. Redistribution metric (only for nostone vs ctrl comparison)
    redistribution_metric = np.nan
    if scenario == "nostone":
        # Use the Gini coefficient computed in Figure 5 or recompute
        ctrl_maxdepth = scenario_data["ctrl"]["max_depth_gdf"].copy()
        nostone_maxdepth = scenario_data["nostone"]["max_depth_gdf"].copy()
        
        ctrl_maxdepth["coord_key"] = list(zip(ctrl_maxdepth.geometry.x, ctrl_maxdepth.geometry.y))
        nostone_maxdepth["coord_key"] = list(zip(nostone_maxdepth.geometry.x, nostone_maxdepth.geometry.y))
        
        depth_merged = nostone_maxdepth.merge(
            ctrl_maxdepth[["coord_key", "max_depth"]],
            on="coord_key",
            how="outer",
            suffixes=("_nostone", "_ctrl")
        )
        depth_merged["max_depth_nostone"] = depth_merged["max_depth_nostone"].fillna(0)
        depth_merged["max_depth_ctrl"] = depth_merged["max_depth_ctrl"].fillna(0)
        depth_merged["depth_diff"] = depth_merged["max_depth_nostone"] - depth_merged["max_depth_ctrl"]
        
        # Redistribution metric: fraction of cells with increased depth
        flooded = depth_merged[(depth_merged["max_depth_nostone"] > 0.01) | (depth_merged["max_depth_ctrl"] > 0.01)]
        n_increase = (flooded["depth_diff"] > 0.01).sum()
        redistribution_metric = n_increase / len(flooded) if len(flooded) > 0 else 0
    
    # Store metrics
    diagnostic_metrics[scenario] = {
        "max_inundated_area_km2": max_inundated_area,
        "area_above_1m_km2": area_above_1m,
        "hours_area_integral": hours_area_integral,
        "presurge_stored_water_Mm3": presurge_stored_water,
        "tide_locking_index": tide_locking_index,
        "redistribution_metric": redistribution_metric,
        "peak_time": peak_time
    }
    
    print(f"  Max inundated area: {max_inundated_area:.1f} km²")
    print(f"  Area > 1m depth: {area_above_1m:.1f} km²")
    print(f"  Hours-area integral: {hours_area_integral:.1f} km²·h")
    print(f"  Pre-surge stored water: {presurge_stored_water:.2f} million m³")
    print(f"  Tide-locking index: {tide_locking_index:.3f}")
    if not np.isnan(redistribution_metric):
        print(f"  Redistribution metric: {redistribution_metric:.3f}")

# Create DataFrame for easier plotting
metrics_df = pd.DataFrame(diagnostic_metrics).T
metrics_df.index.name = "scenario"
print("\n" + "="*60)
print("DIAGNOSTIC METRICS SUMMARY TABLE")
print("="*60)
print(metrics_df.to_string())

# ============================================================================
# FIGURE 6: THREE-PANEL DIAGNOSTIC VISUALIZATION
# ============================================================================
print("\n" + "="*60)
print("GENERATING FIGURE 6")
print("="*60)

fig = plt.figure(figsize=(16, 5), dpi=150)
gs = gridspec.GridSpec(1, 3, width_ratios=[1, 1, 1], wspace=0.3)

# Scenario display names and colors
scenario_names = {
    "ctrl": "Control",
    "lowtide": "Neap Tide",
    "norain1": "No Rain 1",
    "norain2": "No Rain 2",
    "nostone": "No Stone"
}
scenario_colors = {
    "ctrl": "#1f77b4",
    "lowtide": "#ff7f0e",
    "norain1": "#2ca02c",
    "norain2": "#9467bd",
    "nostone": "#d62728"
}

# ============================================================================
# PANEL 6A: RADAR PLOT
# ============================================================================
ax_radar = fig.add_subplot(gs[0], projection="polar")

# Metrics for radar (exclude redistribution since it's only for one scenario)
radar_metrics = ["max_inundated_area_km2", "area_above_1m_km2", "hours_area_integral", 
                 "presurge_stored_water_Mm3", "tide_locking_index"]
radar_labels = ["Max Area\n(km²)", "Area >1m\n(km²)", "Persistence\n(km²·h)", 
                "Pre-surge\nWater (Mm³)", "Tide-lock\nIndex"]

# Normalize metrics to 0-1 for radar plot
radar_data = {}
for metric in radar_metrics:
    values = metrics_df[metric].values
    min_val, max_val = values.min(), values.max()
    if max_val > min_val:
        radar_data[metric] = (values - min_val) / (max_val - min_val)
    else:
        radar_data[metric] = np.ones_like(values) * 0.5

# Number of variables
N = len(radar_metrics)
angles = np.linspace(0, 2 * np.pi, N, endpoint=False).tolist()
angles += angles[:1]  # Complete the loop

# Plot each scenario
for i, scenario in enumerate(metrics_df.index):
    values = [radar_data[m][i] for m in radar_metrics]
    values += values[:1]  # Complete the loop
    
    ax_radar.plot(angles, values, "o-", linewidth=2, 
                  label=scenario_names.get(scenario, scenario),
                  color=scenario_colors.get(scenario, f"C{i}"))
    ax_radar.fill(angles, values, alpha=0.1, color=scenario_colors.get(scenario, f"C{i}"))

ax_radar.set_xticks(angles[:-1])
ax_radar.set_xticklabels(radar_labels, size=8)
ax_radar.set_ylim(0, 1.1)
ax_radar.legend(loc="upper right", bbox_to_anchor=(1.3, 1.1), fontsize=8)
ax_radar.set_title("A. Diagnostic Radar Plot\n(normalized metrics)", fontsize=10, fontweight="bold", pad=20)

# ============================================================================
# PANEL 6B: STACKED BARS
# ============================================================================
ax_bars = fig.add_subplot(gs[1])

# Select key metrics for stacked bars
bar_metrics = ["area_above_1m_km2", "presurge_stored_water_Mm3"]
bar_labels = ["Area >1m (km²)", "Pre-surge Water (Mm³ × 10)"]

x = np.arange(len(metrics_df.index))
width = 0.6

# Normalize for stacking visualization (scale pre-surge water to similar range)
area_1m = metrics_df["area_above_1m_km2"].values
presurge_scaled = metrics_df["presurge_stored_water_Mm3"].values * 10  # Scale for visibility

bars1 = ax_bars.bar(x, area_1m, width, label="Area >1m (km²)", 
                    color=[scenario_colors.get(s, "gray") for s in metrics_df.index], alpha=0.8)
bars2 = ax_bars.bar(x, presurge_scaled, width, bottom=area_1m, 
                    label="Pre-surge Water (Mm³ × 10)", 
                    color=[scenario_colors.get(s, "gray") for s in metrics_df.index], alpha=0.4,
                    hatch="//")

ax_bars.set_ylabel("Combined metric value", fontsize=10)
ax_bars.set_xticks(x)
ax_bars.set_xticklabels([scenario_names.get(s, s) for s in metrics_df.index], rotation=45, ha="right", fontsize=9)
ax_bars.legend(loc="upper right", fontsize=8)
ax_bars.set_title("B. Key Metrics by Scenario\n(stacked bars)", fontsize=10, fontweight="bold")

# Add value labels
for i, (a1, pw) in enumerate(zip(area_1m, presurge_scaled)):
    ax_bars.text(i, a1/2, f"{a1:.0f}", ha="center", va="center", fontsize=7, color="white", fontweight="bold")
    ax_bars.text(i, a1 + pw/2, f"{pw/10:.1f}", ha="center", va="center", fontsize=7, fontweight="bold")

# ============================================================================
# PANEL 6C: DIAGNOSTIC SPACE SCATTER
# ============================================================================
ax_scatter = fig.add_subplot(gs[2])

# X-axis: Preconditioning (pre-surge stored water)
# Y-axis: Tide-locking index
# Marker size: Persistence (hours-area integral)

x_vals = metrics_df["presurge_stored_water_Mm3"].values
y_vals = metrics_df["tide_locking_index"].values
size_vals = metrics_df["hours_area_integral"].values

# Normalize size for visualization
size_min, size_max = size_vals.min(), size_vals.max()
if size_max > size_min:
    sizes = 100 + 400 * (size_vals - size_min) / (size_max - size_min)
else:
    sizes = np.ones_like(size_vals) * 250

for i, scenario in enumerate(metrics_df.index):
    ax_scatter.scatter(x_vals[i], y_vals[i], s=sizes[i], 
                       c=scenario_colors.get(scenario, f"C{i}"),
                       label=scenario_names.get(scenario, scenario),
                       alpha=0.7, edgecolors="black", linewidth=1)

ax_scatter.set_xlabel("Preconditioning\n(Pre-surge stored water, million m³)", fontsize=10)
ax_scatter.set_ylabel("Tide-locking Index", fontsize=10)
ax_scatter.legend(loc="upper left", fontsize=8, title="Scenario", title_fontsize=9)
ax_scatter.set_title("C. Diagnostic Space\n(size = persistence)", fontsize=10, fontweight="bold")

# Add grid
ax_scatter.grid(True, alpha=0.3)
ax_scatter.set_xlim(left=0)
ax_scatter.set_ylim(0, 1.05)

# Add size legend
size_legend_vals = [size_min, (size_min + size_max)/2, size_max]
size_legend_sizes = [100, 300, 500]
for val, sz in zip(size_legend_vals, size_legend_sizes):
    ax_scatter.scatter([], [], s=sz, c="gray", alpha=0.5, 
                       label=f"{val:.0f} km²·h")

# ============================================================================
# FINAL ADJUSTMENTS
# ============================================================================
plt.suptitle("Figure 6: Compact Diagnostic Summary Across Experiments",
             fontsize=12, fontweight="bold", y=1.02)
plt.tight_layout()

fig.savefig(f"{base_dir}/Fig6_Diagnostics_NCities.pdf", dpi=300, bbox_inches="tight")
fig.savefig(f"{base_dir}/Fig6_Diagnostics_NCities.png", dpi=300, bbox_inches="tight")

plt.show()

# ============================================================================
# SUMMARY TABLE OUTPUT
# ============================================================================
print("\n" + "="*60)
print("DIAGNOSTIC METRICS SUMMARY")
print("="*60)
print("\nMetric definitions:")
print("  - Max inundated area: Peak flooded area during simulation (km²)")
print("  - Area >1m: Area with max depth exceeding 1 meter (km²)")
print("  - Hours-area integral: Cumulative area×time (persistence proxy, km²·h)")
print("  - Pre-surge water: Stored water before surge peak (million m³)")
print("  - Tide-locking index: 1=peak at high tide, 0=peak at low tide")
print("  - Redistribution: Fraction of cells with increased depth (nostone only)")

print("\nFigure 6 saved to:")
print(f"  - {base_dir}/Fig6_Diagnostics_NCities.pdf")
print(f"  - {base_dir}/Fig6_Diagnostics_NCities.png")
```
